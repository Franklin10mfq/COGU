Guidance problem to solve:

$$
\begin{align*}
		\min \quad
		&
		\alpha_b\sum_{k=0}^{T-1}\|\bm{b}_{k}\|_1
		\\[6pt]
		\text{s.t.}\quad
		&
		\dot{\bm{q}}=
		\frac{1}{2}
		\begin{bmatrix}
			-q_1&-q_2&-q_3\\
			q_0&-q_3&q_2\\
			q_3&q_0&-q_1\\
			-q_2&q_1&q_0
		\end{bmatrix}
		(\bm{I}^{-1}\bm{h}_B)
		\\[6pt]
		&
		\dot{\bm{h}}_\mathcal{B}
		=
		-(\bm{I}^{-1}\bm{h}_\mathcal{B})
		\times
		\bm{h}_\mathcal{B}
		+
		\sum_{i=1}^{4}\left(
		\bm{r}_i\times I_{\text{bit}_i}
		\sum_{k=0}^{T-1}
		\left(
		b_{i,k}
		\cdot
		\sum_{m=0}^{l}
		\delta\left(
		t-\frac{m}{l}\frac{t_f}{T}k
		\right)
		\right)
		\hat{\bm{f}}_i
		\right)
		\\[6pt]
		&
		0\leq b_{i,k}\leq1,
		\qquad
		b_{1,k}+b_{2,k}\leq1,
		\qquad
		b_{3,k}+b_{4,k}\leq1
		\\[4pt]
		&
		\|J^{-1}\bm{h}_\mathcal{B}\|_2
		\leq
		\bm{\omega}_{\max}
		\\[6pt]
		&
		\bm{q}(0)=\bm{q}_0,
		\quad
		\bm{h}_\mathcal{B}(0)=\bm{h}_{\mathcal{B},0},
		\quad
		\bm{q}(t_f)=\bm{q}_f,
		\quad
		\bm{h}_\mathcal{B}(t_f)=\bm{h}_{\mathcal{B},t_f}.
	\end{align*}

$$

Given the scaling matrices and vectors $\bm{S}_x$, $\bm{S}_u$, $\bm{c}_x$, $\bm{c}_u$:

$$\hat{\bm{u}}=\begin{bmatrix}
\bm{a}\\
\bm{\tau}
\end{bmatrix},\quad\hat{\bm{u}}=\bm{S}_u\bm{u}+\bm{c}_u$$

$$\hat{\bm{x}}=\begin{bmatrix}
\bm{r}\\
\bm{v}\\
\bm{q}\\
\bm{\omega}
\end{bmatrix},\quad\hat{\bm{x}}=\bm{S}_x\bm{x}+\bm{c}_x$$

Where scaling matrices and vectors try to limit the optimization variables within $\bm{x}\in[0, 1]^n$, $\bm{u}\in[0, 1]^m$.

First, the original optimization problem is modeled as follows:
$$
\begin{align*}
    \text{min} \quad &J_\lambda(\bm{x},\bm{u},\bm{v}_c,\bm{v}_i)\\
    \text{s.t.}\quad&
        \dot{\hat{\bm{x}}}=
        \bm{f}(\hat{\bm{x}},\hat{\bm{u}},t)\\
    &\bm{g}(\hat{\bm{x}},\hat{\bm{u}},t)\leq \bm{0}\\
    &||\bm{x}-\bar{\bm{x}}||_\infty+||\bm{u}-\bar{\bm{u}}||_\infty\leq \eta\\
    &\hat{\bm{x}}\in\mathcal{X},\quad \hat{\bm{u}}\in\mathcal{U}
\end{align*}
$$
Where $\mathcal{X}$, $\mathcal{U}$ encompass convex restrictions.

Via 1st order Taylor series and matrix algebra:

$$
\begin{equation*}
\hat{\bm{A}}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)=\nabla_{\bm{\hat{x}}}\bm{f}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t),\quad
\hat{\bm{B}}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)=\nabla_{\bm{\hat{u}}}\bm{f}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)
\end{equation*}
$$
$$
\begin{equation*}
\hat{\bm{y}}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)=\bm{f}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)-\nabla_{\bm{\hat{x}}}\bm{f}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)\hat{\bar{\bm{x}}}-\nabla_{\bm{\hat{u}}}\bm{f}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)\hat{\bar{\bm{u}}}
\end{equation*}
$$

$$
\begin{equation*}
\hat{\bm{C}}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)=\nabla_{\bm{\hat{x}}}\bm{g}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t),\quad
\hat{\bm{D}}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)=\nabla_{\bm{\hat{u}}}\bm{g}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)
\end{equation*}
$$
$$
\begin{equation*}
\hat{\bm{z}}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)=\bm{g}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)-\nabla_{\bm{\hat{x}}}\bm{g}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)\hat{\bar{\bm{x}}}-\nabla_{\bm{\hat{u}}}\bm{g}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)\hat{\bar{\bm{u}}}
\end{equation*}
$$

Then, original optimization problem is transformed into:

$$
\begin{align*}
    \text{min} \quad &J_\lambda(\bm{x},\bm{u},\bm{v}_c,\bm{v}_i)\\
    \text{s.t.}\quad&
        \dot{\hat{\bm{x}}}=
        \hat{\bm{A}}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)\hat{\bm{x}}+\hat{\bm{B}}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)\hat{\bm{u}}+\hat{\bm{y}}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)+\bm{v}_c\\
    &\hat{\bm{C}}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)\hat{\bm{x}}+\hat{\bm{D}}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)\hat{\bm{u}}+\hat{\bm{z}}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)\leq \bm{v}_i\\
    &||\bm{x}-\bar{\bm{x}}||_\infty+||\bm{u}-\bar{\bm{u}}||_\infty\leq \eta\\
    &\hat{\bm{x}}\in\mathcal{X},\quad \hat{\bm{u}}\in\mathcal{U}
\end{align*}
$$

Defining:

$$
\begin{equation*}
\bm{A}=\bm{S}_x^{-1}\hat{\bm{A}}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)\bm{S}_x,\quad \bm{B}=\bm{S}_x^{-1}\hat{\bm{B}}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)\bm{S}_u
\end{equation*}
$$
$$
\begin{equation*}
\bm{y}=\bm{S}_x^{-1}\hat{\bm{A}}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)\bm{c}_x+
\bm{S}_x^{-1}\hat{\bm{B}}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)\bm{c}_u+
\bm{S}_x^{-1}\hat{\bm{y}}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)
\end{equation*}
$$

$$
\begin{equation*}
\bm{C}=\hat{\bm{C}}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)\bm{S}_x,\quad \bm{D}=\hat{\bm{D}}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)\bm{S}_u
\end{equation*}
$$
$$
\begin{equation*}
\bm{z}=\hat{\bm{C}}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)\bm{c}_x+
\hat{\bm{D}}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)\bm{c}_u+
\hat{\bm{z}}(\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}},t)
\end{equation*}
$$

Some matrix algebra leads to the following optimization problem (we omit $\hat{\bar{\bm{x}}},\hat{\bar{\bm{u}}}$ dependency for notation simplicity):
$$
\begin{align*}
    \text{min} \quad &J_\lambda(\bm{x},\bm{u},\bm{v}_c,\bm{v}_i)\\
    \text{s.t.}\quad&
        \dot{\bm{x}}=
        \bm{A}(t)\bm{x}+\bm{B}(t)\bm{u}+\bm{y}(t)+\bm{v}_c\\
    &\bm{C}(t)\bm{x}+\bm{D}(t)\bm{u}+\bm{z}(t)\leq \bm{v}_i\\
    &||\bm{x}-\bar{\bm{x}}||_\infty+||\bm{u}-\bar{\bm{u}}||_\infty\leq \eta\\
    &\bm{x}\in\mathcal{X},\quad \bm{u}\in\mathcal{U}
\end{align*}
$$

Let the Linear Time Varying system (LTV):

$$
\begin{equation*}
    \dot{\bm{x}}(t)=\bm{A}(t)\bm{x}(t)+\bm{B}(t)\bm{u}(t)+\bm{y}(t)
\end{equation*}
$$

For a specific $t \in [t_k,t_{k+1}]$, the exact solution for $\bm{x}$ is given by:

$$
\begin{equation*}
    \bm{x}(t_{k+1})=\bm{\Phi}(t_{k+1},t_k)\bm{x}(t_k)+\int_{t_k}^{t_{k+1}}\Phi(t_{k+1},\xi)\bm{B}(\xi)\bm{u}(\xi)d\xi+\int_{t_k}^{t_{k+1}}\Phi(t_{k+1},\xi)\bm{y}(\xi)d\xi
\end{equation*}
$$

State transition matrix $\bm{\Phi}$ properties lead to:
$$
\begin{equation*}
    \bm{x}(t_{k+1})=\bm{\Phi}(t_{k+1},t_k)\bm{x}(t_k)+\bm{\Phi}(t_{k+1},t_k)\int_{t_k}^{t_{k+1}}\bm{\Phi}^{-1}(\xi,t_k)\bm{B}(\xi)\bm{u}(\xi)d\xi+\bm{\Phi}(t_{k+1},t_k)\int_{t_k}^{t_{k+1}}\bm{\Phi}^{-1}(\xi,t_k)\bm{y}(\xi)d\xi
\end{equation*}
$$

Assuming a Zero Order Hold condition for $\bm{u}(t)$ with $t\in[t_k,t_{k+1})$:
$$
\begin{equation*}
    \bm{x}(t_{k+1}):=\bm{A}_k\bm{x}(t_k)+\bm{B}_k\bm{u}(t_k)+\bm{y}_k
\end{equation*}
$$
For simplicity denoted as:
\begin{equation*}
    \bm{x}_{k+1}=\bm{A}_k\bm{x}_k+\bm{B}_k\bm{u}_k+\bm{y}_k
\end{equation*}
Where:
$$
\begin{equation*}
\bm{A}_k:=\bm{\Phi}(t_{k+1},t_k)
\end{equation*}
$$
$$
\begin{equation*}
\bm{B}_k:=\bm{\Phi}(t_{k+1},t_k)\int_{t_k}^{t_{k+1}}\bm{\Phi}^{-1}(\xi,t_k)\bm{B}(\xi)d\xi,\quad
\bm{y}_k:=\bm{\Phi}(t_{k+1},t_k)\int_{t_k}^{t_{k+1}}\bm{\Phi}^{-1}(\xi,t_k)\bm{y}(\xi)d\xi
\end{equation*}
$$

To compute these discrete matrices at each sampling instant, the following differential equation corresponding to the state transition matrix is solved successively for each discretization step:

$$
\begin{equation*}
\dot{\bm{\Phi}}(t,t_0)=\bm{A}(t)\bm{\Phi}(t,t_0)
\end{equation*}
$$

In this manner, the full SCVx convex optimization problem is formulated:

$$
\begin{align*}
    \text{min} \quad &\mathcal{L}_\lambda(\bm{x},\bm{u},\bm{v}_c,\bm{v}_i)\\
    \text{s.t.}\quad&
        \bm{x}_{k+1}=
        \bm{A}_k\bm{x}+\bm{B}_k\bm{u}_k+\bm{y}_k+\bm{v}_{c_k}\\
    &\bm{C}_{k}\bm{x}_{k}+\bm{D}_{k}\bm{u}_{k}+\bm{z}_{k}\leq \bm{v}_{i_k}\\
    &||\bm{x}_{k}-\bar{\bm{x}}_{k}||_1+||\bm{u}_{k}-\bar{\bm{u}}_{k}||_1\leq \eta\\
    &\bm{x}_k\in\mathcal{X}_k,\quad \bm{u}_k\in\mathcal{U}_k
\end{align*}
$$

This example denotes for instance (not always):
- $\hat{\bm{x}}$ as `hx`(no scaled).
- $\bm{x}$ as `x` (scaled)
- $\bm{x}_k$ as `x[:,k:k+1]` (discrete scaled)
- $\bar{\bm{x}}_k$ as `ox[:,k:k+1]` (discrete scaled old, last generated trajectory)

```python
def replace_variables_in_string(f_str,
                                state_parameters_string,
                                control_parameters_string,
                                dynamic_parameters_string):

    # Replace states
    for i, var in enumerate(state_parameters_string):
        f_str = re.sub(rf"\b{var}\b", f"hx[{i},0]", f_str)

    # Replace inputs
    for i, var in enumerate(control_parameters_string):
        f_str = re.sub(rf"\b{var}\b", f"hu[{i},0]", f_str)

    # Replace problem parameters
    for i, var in enumerate(dynamic_parameters_string):
        f_str = re.sub(rf"\b{var}\b", f"dyn_par[{i}]", f_str)

    return f_str

def matrix_to_numpy_string(matrix, state_params, control_params, dynamic_params):
    """
    Converts a sympy symbolic matrix into a string representing a numpy.array,
    with optional variable replacement using replace_variables_in_string.

    Args:
        matrix: sympy.Matrix of any dimension.
        state_params, control_params, dynamic_params: strings for variable replacement.

    Returns:
        A string representing a numpy.array ready to be evaluated in Python.
    """

    n_rows, *rest = np.shape(matrix)
    n_cols = rest[0] if rest else 1
    
    matrix = sp.Matrix(n_rows, n_cols, lambda i, j: matrix[i*n_cols + j])

    array_str = "np.array([["

    for i in range(0,n_rows):
        for j in range(0,n_cols):
            array_str += sp.pycode(matrix[i, j])
            if j < n_cols - 1:
                array_str += ","
        if i < n_rows - 1:
            array_str += "],["
        else:
            array_str += "]])"

    # Apply variable replacement using your custom function
    array_str = replace_variables_in_string(array_str,
                                            state_params,
                                            control_params,
                                            dynamic_params)
    return array_str



#Automated Python script creation

#f
string_f_SCVx=matrix_to_numpy_string(f_orig_opt_prob,
                                                    state_parameters_string,
                                                    control_parameters_string,
                                                    dynamic_parameters_string)
print("\nf_SCVx:\n")
print(string_f_SCVx)

#A, B, y
A_symbolic = sp.Matrix(f_orig_opt_prob).jacobian(x_states) #Sympy Jacobian (Linearization via Taylor series)

string_A_SCVx =string_A_SCVx=matrix_to_numpy_string(A_symbolic,
                                                    state_parameters_string,
                                                    control_parameters_string,
                                                    dynamic_parameters_string)
print("\nA_no_discrete_no_scaled:\n")
print(string_A_SCVx)

B_symbolic = sp.Matrix(f_orig_opt_prob).jacobian(u_input) #Sympy Jacobian (Linearization via Taylor series)

string_B_SCVx =string_B_SCVx=matrix_to_numpy_string(B_symbolic,
                                                    state_parameters_string,
                                                    control_parameters_string,
                                                    dynamic_parameters_string)
print("\nB_no_discrete_no_scaled:\n")
print(string_B_SCVx)

y_symbolic = f_orig_opt_prob-A_symbolic@np.transpose(x_states)-B_symbolic@np.transpose(u_input) #Sympy Jacobian (Linearization via Taylor series)

string_y_SCVx =string_y_SCVx=matrix_to_numpy_string(y_symbolic,
                                                    state_parameters_string,
                                                    control_parameters_string,
                                                    dynamic_parameters_string)
print("\ny_no_discrete_no_scaled:\n")
print(string_y_SCVx)

string_g_SCVx = "0"
string_C_SCVx = "0"
string_D_SCVx = "0"
string_z_SCVx = "0"

if(non_convex_inequalities_size>0):
    #g
    string_g_SCVx=matrix_to_numpy_string(g_orig_opt_prob,
                                                    state_parameters_string,
                                                    control_parameters_string,
                                                    dynamic_parameters_string)
    print("\ng_SCVx:\n")
    print(string_g_SCVx)
    #C, D, z
    C_symbolic = sp.Matrix(g_orig_opt_prob).jacobian(x_states) #Sympy Jacobian (Linearization via Taylor series)

    string_C_SCVx = matrix_to_numpy_string(C_symbolic,
                                            state_parameters_string,
                                            control_parameters_string,
                                            dynamic_parameters_string)
    print("\nC_no_discrete_no_scaled:\n")
    print(string_C_SCVx)

    D_symbolic = sp.Matrix(g_orig_opt_prob).jacobian(u_input) #Sympy Jacobian (Linearization via Taylor series)

    string_D_SCVx = matrix_to_numpy_string(D_symbolic,
                                            state_parameters_string,
                                            control_parameters_string,
                                            dynamic_parameters_string)
    print("\nD_no_discrete_no_scaled:\n")
    print(string_D_SCVx)

    z_symbolic = g_orig_opt_prob-C_symbolic@np.transpose(x_states)-D_symbolic@np.transpose(u_input) #Sympy Jacobian (Linearization via Taylor series)

    string_z_SCVx = matrix_to_numpy_string(z_symbolic,
                                            state_parameters_string,
                                            control_parameters_string,
                                            dynamic_parameters_string)
    print("\nz_no_discrete_no_scaled:\n")
    print(string_z_SCVx)

#Write f, g, A, y, C, D, z functions inside f_g_A_B_y_C_D_z_functions.py

filename = "f_g_A_B_y_C_D_z_functions.py"

file_content = f"""
import numpy as np
import math

def f_SCVx(hx, hu, t, dyn_par):
    return {string_f_SCVx}

def g_SCVx(hx, hu, t, dyn_par):
    return {string_g_SCVx}

def A_no_discrete_no_scaled_SCVx(hx, hu, t, dyn_par):
    return {string_A_SCVx}

def B_no_discrete_no_scaled_SCVx(hx, hu, t, dyn_par):
    return {string_B_SCVx}

def y_no_discrete_no_scaled_SCVx(hx, hu, t, dyn_par):
    return {string_y_SCVx}

def C_no_discrete_no_scaled_SCVx(hx, hu, t, dyn_par):
    return {string_C_SCVx}

def D_no_discrete_no_scaled_SCVx(hx, hu, t, dyn_par):
    return {string_D_SCVx}

def z_no_discrete_no_scaled_SCVx(hx, hu, t, dyn_par):
    return {string_z_SCVx}
"""
with open(filename, "w", encoding="utf-8") as f:
    f.write(file_content)
print("\nSymbolic algebra functions implemented!")
```

**Warm starting**: The user should input this. In particular, this example uses:
- Slerp method to compute the initial quaternion sequence, then via Rigid body kinematics, the angular velocity sequence is computed. 
- Linear interpolation to compute the initial velocity sequence, then via discrete derivative, the velocity sequence is computed.

```python
def slerp(q1, q2, num_samples):
    dot = np.dot(q1, q2)
    if dot < 0.0:
        q2 = -q2
        dot = -dot
    dot = np.clip(dot, -1.0, 1.0)
    theta_0 = np.arccos(dot)

    if np.abs(theta_0) < 1e-6:
        return np.linspace(q1, q2, num_samples)

    sin_theta_0 = np.sin(theta_0)

    quaternions = []
    for i in range(num_samples):
        t = i / (num_samples - 1)
        theta = theta_0 * t
        sin_theta = np.sin(theta)
        s0 = np.cos(theta) - dot * sin_theta / sin_theta_0
        s1 = sin_theta / sin_theta_0
        quaternions.append(s0 * q1 + s1 * q2)

    return np.array(quaternions)

def compute_angular_velocity(quaternions, dt):

    rotations = R.from_quat(quaternions)
    angular_velocities = [[0,0,0]]

    for i in range(len(rotations) - 1):
        delta_rot = rotations[i + 1] * rotations[i].inv()
        log_rot = delta_rot.as_rotvec() / dt
        angular_velocities.append(log_rot)

    return np.array(angular_velocities)

# Initial trajectory (without scaling)

ohx_quat = slerp(start_pos_double[0:4,0], end_pos_double[0:4,0],T+1).T
ohx_hb = Jsat_double@compute_angular_velocity(ohx_quat.T, tau_double).T

ohx_init=np.zeros((states_size,T+1))
ohx_init[0:4,:]=ohx_quat
ohx_init[4:7,:]=ohx_hb

ohu_init=np.zeros((inputs_size,T))
```

```python
def SCVx_scaling_x(x_no_scaled,inv_S_x_scaling_double,c_x_scaling_double):
    x_scaled = inv_S_x_scaling_double@(x_no_scaled-c_x_scaling_double)
    return x_scaled
    
def SCVx_scaling_u(u_no_scaled,inv_S_u_scaling_double,c_u_scaling_double):
    u_scaled = inv_S_u_scaling_double@(u_no_scaled-c_u_scaling_double)
    return u_scaled

def SCVx_inv_scaling_x(x_scaled,S_x_scaling_double,c_x_scaling_double):
    x_inv_scaling = S_x_scaling_double@x_scaled+c_x_scaling_double
    return x_inv_scaling
    
def SCVx_inv_scaling_u(u_scaled,S_u_scaling_double,c_u_scaling_double):
    u_inv_scaling = S_u_scaling_double@u_scaled+c_u_scaling_double
    return u_inv_scaling


def A_no_discrete_scaled_SCVx(A_no_discrete_no_scaled, inv_S_x_scaling_double,S_x_scaling_double):

    A_numeric = inv_S_x_scaling_double@A_no_discrete_no_scaled@S_x_scaling_double

    return A_numeric

def B_no_discrete_scaled_SCVx(B_no_discrete_no_scaled, inv_S_x_scaling_double, S_u_scaling_double):

    B_numeric = inv_S_x_scaling_double@B_no_discrete_no_scaled@S_u_scaling_double

    return B_numeric

def y_no_discrete_scaled_SCVx(A_no_discrete_no_scaled,
                              B_no_discrete_no_scaled,
                              y_no_discrete_no_scaled,
                              inv_S_x_scaling_double, c_x_scaling_double, c_u_scaling_double):

    y_numeric = inv_S_x_scaling_double@A_no_discrete_no_scaled@c_x_scaling_double
    y_numeric = y_numeric+inv_S_x_scaling_double@B_no_discrete_no_scaled@c_u_scaling_double
    y_numeric = y_numeric+inv_S_x_scaling_double@y_no_discrete_no_scaled

    return y_numeric
    
def A_B_y_discrete_scaled(ohx, ohu, t0, dynamic_parameters_double, tau_double,
                          S_x_scaling_double, inv_S_x_scaling_double, c_x_scaling_double,
                          S_u_scaling_double, inv_S_u_scaling_double, c_u_scaling_double, size_N):

    n = len(ohx)
    m = len(ohu)

    Phi = np.eye(n)
    G   = np.zeros((n, m))
    z   = np.zeros((n, 1))

    dt = tau_double / (size_N - 1)
    t  = t0

    for i in range(size_N - 1):

        # ===== k1 =====
        A1_no = A_no_discrete_no_scaled_SCVx(ohx, ohu, t, dynamic_parameters_double)
        B1_no = B_no_discrete_no_scaled_SCVx(ohx, ohu, t, dynamic_parameters_double)
        y1_no = y_no_discrete_no_scaled_SCVx(ohx, ohu, t, dynamic_parameters_double)

        A1 = A_no_discrete_scaled_SCVx(A1_no, inv_S_x_scaling_double, S_x_scaling_double)
        B1 = B_no_discrete_scaled_SCVx(B1_no, inv_S_x_scaling_double, S_u_scaling_double)
        y1 = y_no_discrete_scaled_SCVx(A1_no, B1_no, y1_no,
                                       inv_S_x_scaling_double,
                                       c_x_scaling_double,
                                       c_u_scaling_double)

        k1_Phi = A1 @ Phi
        k1_G   = A1 @ G + B1
        k1_z   = A1 @ z + y1

        # ===== k2 =====
        t2 = t + 0.5*dt
        Phi2 = Phi + 0.5*dt*k1_Phi
        G2   = G   + 0.5*dt*k1_G
        z2   = z   + 0.5*dt*k1_z

        A2_no = A_no_discrete_no_scaled_SCVx(ohx, ohu, t2, dynamic_parameters_double)
        B2_no = B_no_discrete_no_scaled_SCVx(ohx, ohu, t2, dynamic_parameters_double)
        y2_no = y_no_discrete_no_scaled_SCVx(ohx, ohu, t2, dynamic_parameters_double)

        A2 = A_no_discrete_scaled_SCVx(A2_no, inv_S_x_scaling_double, S_x_scaling_double)
        B2 = B_no_discrete_scaled_SCVx(B2_no, inv_S_x_scaling_double, S_u_scaling_double)
        y2 = y_no_discrete_scaled_SCVx(A2_no, B2_no, y2_no,
                                       inv_S_x_scaling_double,
                                       c_x_scaling_double,
                                       c_u_scaling_double)

        k2_Phi = A2 @ Phi2
        k2_G   = A2 @ G2 + B2
        k2_z   = A2 @ z2 + y2

        # ===== k3 =====
        Phi3 = Phi + 0.5*dt*k2_Phi
        G3   = G   + 0.5*dt*k2_G
        z3   = z   + 0.5*dt*k2_z

        A3_no = A_no_discrete_no_scaled_SCVx(ohx, ohu, t2, dynamic_parameters_double)
        B3_no = B_no_discrete_no_scaled_SCVx(ohx, ohu, t2, dynamic_parameters_double)
        y3_no = y_no_discrete_no_scaled_SCVx(ohx, ohu, t2, dynamic_parameters_double)

        A3 = A_no_discrete_scaled_SCVx(A3_no, inv_S_x_scaling_double, S_x_scaling_double)
        B3 = B_no_discrete_scaled_SCVx(B3_no, inv_S_x_scaling_double, S_u_scaling_double)
        y3 = y_no_discrete_scaled_SCVx(A3_no, B3_no, y3_no,
                                       inv_S_x_scaling_double,
                                       c_x_scaling_double,
                                       c_u_scaling_double)

        k3_Phi = A3 @ Phi3
        k3_G   = A3 @ G3 + B3
        k3_z   = A3 @ z3 + y3

        # ===== k4 =====
        t4 = t + dt
        Phi4 = Phi + dt*k3_Phi
        G4   = G   + dt*k3_G
        z4   = z   + dt*k3_z

        A4_no = A_no_discrete_no_scaled_SCVx(ohx, ohu, t4, dynamic_parameters_double)
        B4_no = B_no_discrete_no_scaled_SCVx(ohx, ohu, t4, dynamic_parameters_double)
        y4_no = y_no_discrete_no_scaled_SCVx(ohx, ohu, t4, dynamic_parameters_double)

        A4 = A_no_discrete_scaled_SCVx(A4_no, inv_S_x_scaling_double, S_x_scaling_double)
        B4 = B_no_discrete_scaled_SCVx(B4_no, inv_S_x_scaling_double, S_u_scaling_double)
        y4 = y_no_discrete_scaled_SCVx(A4_no, B4_no, y4_no,
                                       inv_S_x_scaling_double,
                                       c_x_scaling_double,
                                       c_u_scaling_double)

        k4_Phi = A4 @ Phi4
        k4_G   = A4 @ G4 + B4
        k4_z   = A4 @ z4 + y4

        # ===== Update =====
        Phi += (dt/6)*(k1_Phi + 2*k2_Phi + 2*k3_Phi + k4_Phi)
        G   += (dt/6)*(k1_G   + 2*k2_G   + 2*k3_G   + k4_G)
        z   += (dt/6)*(k1_z   + 2*k2_z   + 2*k3_z   + k4_z)

        t += dt

    return Phi, G, z

def C_no_discrete_scaled_SCVx(C_no_discrete_no_scaled, S_x_scaling_double):

    C_numeric = C_no_discrete_no_scaled@S_x_scaling_double

    return C_numeric

def D_no_discrete_scaled_SCVx(D_no_discrete_no_scaled, S_u_scaling_double):

    D_numeric = D_no_discrete_no_scaled@S_u_scaling_double

    return D_numeric

def z_no_discrete_scaled_SCVx(C_no_discrete_no_scaled,
                              D_no_discrete_no_scaled,
                              z_no_discrete_no_scaled,
                              c_x_scaling_double, c_u_scaling_double):

    z_numeric = C_no_discrete_no_scaled@c_x_scaling_double
    z_numeric = z_numeric+D_no_discrete_no_scaled@c_u_scaling_double
    z_numeric = z_numeric+z_no_discrete_no_scaled

    return z_numeric

def C_D_z_discrete_scaled(ohx, ohu, t0, dynamic_parameters,
                           S_x_scaling_double, c_x_scaling_double,
                           S_u_scaling_double, c_u_scaling_double):
    

    C_no_discrete_no_scaled = C_no_discrete_no_scaled_SCVx(ohx, ohu, t0, dynamic_parameters)
    C_no_discrete_scaled = C_no_discrete_scaled_SCVx(C_no_discrete_no_scaled, S_x_scaling_double)

    
    D_no_discrete_no_scaled = D_no_discrete_no_scaled_SCVx(ohx, ohu, t0, dynamic_parameters)
    D_no_discrete_scaled = D_no_discrete_scaled_SCVx(D_no_discrete_no_scaled, S_u_scaling_double)

    z_no_discrete_no_scaled = z_no_discrete_no_scaled_SCVx(ohx, ohu, t0, dynamic_parameters)
    z_no_discrete_scaled = z_no_discrete_scaled_SCVx(C_no_discrete_no_scaled,
                                                     D_no_discrete_no_scaled,
                                                     z_no_discrete_no_scaled,
                                                     c_x_scaling_double, c_u_scaling_double)
    
    #Discretization no needed, bypass
    return C_no_discrete_scaled, D_no_discrete_scaled, z_no_discrete_scaled

def f_rk4_step(xk, uk, t, dt, dynamic_parameters,
               S_x_scaling_double,c_x_scaling_double,S_u_scaling_double,c_u_scaling_double):
    
    xk_no_scaled = SCVx_inv_scaling_x(xk,S_x_scaling_double,c_x_scaling_double)
    uk_no_scaled = SCVx_inv_scaling_u(uk,S_u_scaling_double,c_u_scaling_double)
    k1 = f_SCVx(xk_no_scaled, uk_no_scaled, t, dynamic_parameters)
    k2 = f_SCVx(xk_no_scaled + 0.5 * dt * k1, uk_no_scaled, t + 0.5*dt, dynamic_parameters)
    k3 = f_SCVx(xk_no_scaled + 0.5 * dt * k2, uk_no_scaled, t + 0.5*dt, dynamic_parameters)
    k4 = f_SCVx(xk_no_scaled + dt * k3, uk_no_scaled, t + dt, dynamic_parameters)
    return xk_no_scaled + (dt / 6) * (k1 + 2 * k2 + 2 * k3 + k4)

def f_Euler_step(xk, uk, t, dt, dynamic_parameters,
                 S_x_scaling_double,c_x_scaling_double,S_u_scaling_double,c_u_scaling_double):

    xk_no_scaled = SCVx_inv_scaling_x(xk,S_x_scaling_double,c_x_scaling_double)
    uk_no_scaled = SCVx_inv_scaling_u(uk,S_u_scaling_double,c_u_scaling_double)

    return xk_no_scaled + dt * f_SCVx(xk_no_scaled, uk_no_scaled, t, dynamic_parameters)
```

Modify the SCVx cost function according to user requirements:

```python
def J_SCVx(x,u,T,tau_double, dynamic_parameters, xq_goal_double,
           S_x_scaling_double,c_x_scaling_double,S_u_scaling_double,c_u_scaling_double):
    
    cost = 0
    for k in range(0, T): # from 0 to T-1
        cost += 5*tau_double*np.linalg.norm(u[0:4,k:k+1], ord=1)

    #for k in range(0, T+1):
    #    cost += 1*tau_double*np.linalg.norm(x[0:7,k:k+1]-xq_goal_double, ord=2)**2

    for k in range(0, T): # from 0 to T-1

        flow_map = f_rk4_step(x[0:, k:k+1], u[:, k:k+1], k/T*tf, tau_double, dynamic_parameters,
        S_x_scaling_double,c_x_scaling_double,S_u_scaling_double,c_u_scaling_double)

        SCVx_defect = SCVx_inv_scaling_x(x[:, k+1:k+2],S_x_scaling_double,c_x_scaling_double)-flow_map

        cost += tau_double*np.linalg.norm(lamb_double*SCVx_defect, ord=1)

    for k in range(0, T+1): # from 0 to T
        x_no_scaled = SCVx_inv_scaling_x(x[:,k:k+1],S_x_scaling_double,c_x_scaling_double)
        if k == T:
            u_no_scaled = SCVx_inv_scaling_u(u[:,T-1:T],S_u_scaling_double,c_u_scaling_double)
        else:
            u_no_scaled = SCVx_inv_scaling_u(u[:,k:k+1],S_u_scaling_double,c_u_scaling_double)
        if non_convex_inequalities_size>0:    
            g_double = g_SCVx(x_no_scaled, u_no_scaled, k/T*tf, dynamic_parameters_double)

            for i in range(0,np.size(g_double)):
                cost += tau_double*np.abs(lamb_double*(np.max([g_double[i,0],0])))

    return cost
```

```python
B = np.column_stack([
    np.cross(rth1_double[:, 0], fth1_double[:, 0]),
    np.cross(rth2_double[:, 0], fth2_double[:, 0]),
    np.cross(rth3_double[:, 0], fth3_double[:, 0]),
    np.cross(rth4_double[:, 0], fth4_double[:, 0]),
])
 
 
# =========================================
# Control interpolation
# =========================================
def u_interpol_zoh(t, u_value, dt_control):
    """Zero-order hold: the SCVx control is constant over each interval."""
    T_control = u_value.shape[1]
    index = int(np.floor(t / dt_control))
    index = min(index, T_control - 1)
    return u_value[:, index]

# =========================================
# Dynamics
# =========================================
def dynamics(t, x, u, params):
 
    q = x[0:4]
    w = x[4:7]
 
    I     = params["I"]
    I_inv = params["I_inv"]
    B     = params["B"]
 
    torque = B @ u                                   # M = sum_i u_i (rth_i x fth_i)
    w_dot  = I_inv @ (torque - np.cross(w, I @ w))   # Euler
 
    wx, wy, wz = w
    q0, q1, q2, q3 = q
 
    q_dot = 0.5 * np.array([
        -wx*q1 - wy*q2 - wz*q3,
         wx*q0 + wz*q2 - wy*q3,
         wy*q0 - wz*q1 + wx*q3,
         wz*q0 + wy*q1 - wx*q2
    ])
 
    return np.concatenate((q_dot, w_dot))
 
 
# =========================================
# RK4 Integrator
# =========================================
def rk4_step(f, t, x, u, dt, params):
 
    k1 = f(t,        x,             u, params)
    k2 = f(t + dt/2, x + dt/2 * k1, u, params)
    k3 = f(t + dt/2, x + dt/2 * k2, u, params)
    k4 = f(t + dt,   x + dt   * k3, u, params)
 
    return x + dt/6 * (k1 + 2*k2 + 2*k3 + k4)
 
 
# =========================================
# Simulation setup
# =========================================
I = Jsat_double.copy()
 
params = {
    "I":     I,
    "I_inv": np.linalg.inv(I),
    "B":     B,
}
 
# Initial state = guidance node 0.  x_value[4:7] is already omega [rad/s].
x = x_value[:, 0].copy()
 
dt_state   = 0.1                          # integration step [s]
dt_control = tau_double                   # guidance step (ZOH) [s]
sim_time   = tf                           # guidance horizon = T*tau
N = int(round(sim_time / dt_state)) + 1
 
if abs(dt_control / dt_state - round(dt_control / dt_state)) > 1e-9:
    print(f"[warning] dt_state={dt_state}s does not divide dt_control={dt_control}s "
          f"-> ZOH boundaries do not land exactly on integration steps")
 
q_history = np.zeros((N, 4))
w_history = np.zeros((N, 3))
u_history = np.zeros((N, 4))
t_history = np.zeros(N)
 
t = 0.0
 
# =========================================
# Simulation loop  (open-loop: SCVx ZOH control)
# =========================================
for i in range(N):
 
    q_history[i] = x[0:4]
    w_history[i] = x[4:7]
    t_history[i] = t
 
    u_current = u_interpol_zoh(t, 1.0*u_value, dt_control)
    u_history[i] = u_current
 
    x = rk4_step(dynamics, t, x, u_current, dt_state, params)
    x[0:4] /= np.linalg.norm(x[0:4])      # renormalize q
    t += dt_state
 
 
# =========================================
# Plots
# =========================================
t_nodes = np.arange(T + 1) * tau_double
 
plt.figure()
plt.plot(t_history, q_history)
plt.plot(t_history, np.linalg.norm(q_history, axis=1))
plt.plot(t_nodes, x_value[0:4, :].T, 'o', ms=3, alpha=0.5)
plt.legend(["q0","q1","q2","q3","|q|"])
plt.title("Quaternion Evolution  (- open-loop, o guidance)")
plt.xlabel("Time [s]"); plt.grid(); plt.show()
 
plt.figure()
plt.plot(t_history, w_history * 180/np.pi)
plt.plot(t_history, np.linalg.norm(w_history, axis=1) * 180/np.pi)
plt.plot(t_nodes, x_value[4:7, :].T * 180/np.pi, 'o', ms=3, alpha=0.5)
plt.legend(["wx","wy","wz","|w|"])
plt.title("Angular Velocity [deg/s]  (- open-loop, o guidance)")
plt.xlabel("Time [s]"); plt.grid(); plt.show()

fig, axs = plt.subplots(4, 1, sharex=True, figsize=(6, 6))

colors = ['tab:blue', 'tab:orange', 'tab:green', 'tab:red']

for i in range(4):
    axs[i].step(t_history, u_history[:, i] * 1e6, where='post',
                color=colors[i], label=f'VAT{i+1}')
    axs[i].set_ylabel(f'Fth{i+1} [uN]')
    axs[i].grid(True)
    axs[i].legend(loc='upper right')
    axs[i].tick_params(labelsize=10)

axs[-1].set_xlabel('Time [s]')
fig.suptitle('Thruster force sequence [uN]')
fig.tight_layout()
plt.show()
```

```python
# ============================================================================
#  VAT OPEN-LOOP COGU on a PERTURBED PLANT  (no RW)
# ----------------------------------------------------------------------------
#  Reaction wheels removed. The COGU pulse train u_value is applied open-loop
#  to the TRUE perturbed plant. State = [q(4), omega(3)]  (x_value[4:7] is omega).
#  Perturbations:
#     - inertia error (I_true vs I_nom)
#     - actuator noise (offset + per-thruster bias + Gaussian, only while firing)
#     - tau_dist : external disturbance torque, ALWAYS ON (even with VAT off)
# ============================================================================

import numpy as np
import matplotlib.pyplot as plt

# =========================================
# Thruster control matrix B
# =========================================
B = np.column_stack([
    np.cross(rth1_double[:, 0], fth1_double[:, 0]),
    np.cross(rth2_double[:, 0], fth2_double[:, 0]),
    np.cross(rth3_double[:, 0], fth3_double[:, 0]),
    np.cross(rth4_double[:, 0], fth4_double[:, 0]),
])

I_nominal = Jsat_double.copy()

# =========================================
# Perturbations / uncertainty
# =========================================
pert_seed = 0
rng = np.random.default_rng(pert_seed)

# --- actuator noise (only while a thruster fires) ---
thrust_nominal     = Fth_max_double
thrust_offset      = 0.0                    # systematic offset on every firing [N]   (TUNE)
thrust_sigma       = Fth_max_double / 10    # per-firing Gaussian std [N]
thrust_bias_sigma  = Fth_max_double / 20    # fixed thruster-to-thruster bias std [N]
thrust_bias        = thrust_bias_sigma * rng.standard_normal(4)

# --- inertia uncertainty ---
inertia_var = 0.05                         # +/- 5% (set 0.0 to disable)             (TUNE)

# --- EXTERNAL disturbance torque: ALWAYS applied, even when VAT = 0 ---

def disturbance_torque(t):
    return np.array([
        1.0e-7*np.sin(2*np.pi*t/5400),
        0.8e-7*np.sin(2*np.pi*t/4200 + 1.3),
        0.6e-7*np.sin(2*np.pi*t/6800 + 2.1)
    ])

def _deficit_distribute(counts, n):
    """
    Reparte n slots [0..n-1] entre len(counts) categorías, de forma
    determinística y lo más uniforme posible, según target = counts[c].
    sum(counts) debe ser <= n (el resto queda implícito si counts no suma n).
    Devuelve una lista de longitud n con el índice de categoría por slot.

    Algoritmo: en cada paso i, se asigna el slot a la categoría con mayor
    "déficit" = target_fraccional_acumulado - ya_asignado. Generalización
    multi-categoría de Bresenham / Sainte-Laguë sequential apportionment.
    """
    n_cat = len(counts)
    assigned = [0] * n_cat
    out = [-1] * n
    for i in range(n):
        best_c, best_deficit = -1, -1.0
        for c in range(n_cat):
            if counts[c] <= 0:
                continue
            target = counts[c] * (i + 1) / n
            deficit = target - assigned[c]
            if deficit > best_deficit:
                best_deficit, best_c = deficit, c
        if best_c == -1:           # ningún target > 0 (caso todo-cero)
            best_c = n_cat - 1      # cae en categoría "ninguno" si existe
        assigned[best_c] += 1
        out[i] = best_c
    return out


def _pair_fire_pattern(a, b, n_sub, Fmax):
    """
    Para un par antagonista {A,B} (mutuamente excluyentes), devuelve un
    array de longitud n_sub con 0=ninguno, 1=A, 2=B en cada sub-slot,
    distribuido lo más uniformemente posible.
    a, b >= 0 ; a+b <= Fmax (si a+b < Fmax, algunos slots quedan en 0).
    """
    ka = int(round((a / Fmax) * n_sub))
    kb = int(round((b / Fmax) * n_sub))
    ka = min(ka, n_sub)
    kb = min(kb, n_sub - ka)          # garantiza ka+kb <= n_sub
    k_none = n_sub - ka - kb
    # orden de categorías: 0=none, 1=A, 2=B
    pattern = _deficit_distribute([k_none, ka, kb], n_sub)
    return pattern


def u_interpol_zoh(t, u_value, dt_control, n_sub=10):
    T_control = u_value.shape[1]
    index = int(np.floor(t / dt_control))
    if index >= T_control:
        return np.zeros(u_value.shape[0])
    u = u_value[:, index]
    t_local = t - index * dt_control
    dt_sub = dt_control / n_sub
    slot_idx = min(int(np.floor(t_local / dt_sub)), n_sub - 1)

    out = np.zeros(u_value.shape[0])

    # --- Par {VAT1, VAT2} ---
    pattern01 = _pair_fire_pattern(u[0], u[1], n_sub, thrust_nominal)
    cat = pattern01[slot_idx]
    if cat == 1:
        out[0] = thrust_nominal
    elif cat == 2:
        out[1] = thrust_nominal

    # --- Par {VAT3, VAT4} ---
    pattern23 = _pair_fire_pattern(u[2], u[3], n_sub, thrust_nominal)
    cat = pattern23[slot_idx]
    if cat == 1:
        out[2] = thrust_nominal
    elif cat == 2:
        out[3] = thrust_nominal

    return out

def apply_thruster_noise(u_cmd):
    fired = u_cmd > 0.0
    noisy = u_cmd + thrust_offset + thrust_bias + thrust_sigma * rng.standard_normal(4)
    return np.where(fired, np.clip(noisy, 0.0, None), 0.0)

# =========================================
# Dynamics  (state = [q, omega])  with disturbance torque, no wheels
# =========================================
def dynamics(t, x, u, params):
    q = x[0:4]; w = x[4:7]
    I, I_inv = params["I"], params["I_inv"]
    tau_ff = (params["B"] @ u) if params["use_thrusters"] else np.zeros(3)
    tau_d = disturbance_torque(t)
    w_dot = I_inv @ (tau_ff + tau_d - np.cross(w, I @ w))   # + disturbance
    wx, wy, wz = w; q0, q1, q2, q3 = q
    q_dot = 0.5 * np.array([
        -wx*q1 - wy*q2 - wz*q3,
         wx*q0 + wz*q2 - wy*q3,
         wy*q0 - wz*q1 + wx*q3,
         wz*q0 + wy*q1 - wx*q2])
    return np.concatenate((q_dot, w_dot))

def rk4_step(f, t, x, u, dt, params):
    k1 = f(t,        x,             u, params)
    k2 = f(t + dt/2, x + dt/2 * k1, u, params)
    k3 = f(t + dt/2, x + dt/2 * k2, u, params)
    k4 = f(t + dt,   x + dt   * k3, u, params)
    return x + dt/6 * (k1 + 2*k2 + 2*k3 + k4)

# =========================================
# Setup
# =========================================
while True:                                  # TRUE perturbed inertia
    P = 1.0 + inertia_var * (2.0 * rng.random((3, 3)) - 1.0)
    P = 0.5 * (P + P.T)
    I_true = I_nominal * P
    if np.all(np.linalg.eigvalsh(I_true) > 0.0):
        break
print("Nominal inertia diag [kg.m^2]:", np.diag(I_nominal))
print("True    inertia diag [kg.m^2]:", np.diag(I_true))
print("Diagonal inertia error (%):", np.round(100*(np.diag(I_true)-np.diag(I_nominal))/np.diag(I_nominal), 1))

params = {
    "I": I_true, "I_inv": np.linalg.inv(I_true), "B": B,
    "use_thrusters": True,
}

# Initial state: [q0, omega0]
x = x_value[:, 0].copy()
x[0:4] /= np.linalg.norm(x[0:4])

dt_state   = 0.1
dt_control = tau_double
hold_time  = 0*120.0
sim_time   = tf + hold_time
N = int(round(sim_time / dt_state)) + 1

q_history = np.zeros((N, 4)); w_history = np.zeros((N, 3))
u_history = np.zeros((N, 4)); ucmd_history = np.zeros((N, 4))
err_history = np.zeros(N); t_history = np.zeros(N)

def q_ref_at(t):
    idx = min(int(np.floor(t / dt_control)), T - 1)
    a = (t - idx * dt_control) / dt_control
    q0 = x_value[0:4, idx]; q1 = x_value[0:4, idx + 1]
    if np.dot(q0, q1) < 0.0: q1 = -q1
    qr = (1 - a) * q0 + a * q1
    return qr / np.linalg.norm(qr)

t = 0.0
for i in range(N):
    q_history[i] = x[0:4]; w_history[i] = x[4:7]; t_history[i] = t

    qr = q_ref_at(min(t, tf))
    err_history[i] = 2.0 * np.degrees(np.arccos(np.clip(abs(np.dot(x[0:4], qr)), 0, 1)))

    u_cmd = u_interpol_zoh(t, u_value, dt_control)
    u_current = apply_thruster_noise(u_cmd)
    ucmd_history[i] = u_cmd; u_history[i] = u_current

    x = rk4_step(dynamics, t, x, u_current, dt_state, params)
    x[0:4] /= np.linalg.norm(x[0:4])
    t += dt_state

q_end = x_value[0:4, T] / np.linalg.norm(x_value[0:4, T])
final_pt = 2.0 * np.degrees(np.arccos(np.clip(abs(np.dot(q_history[-1], q_end)), 0, 1)))
print(f"\nOpen-loop drift vs COGU reference:")
print(f"  final tracking error       : {err_history[-1]:.3f} deg")
print(f"  max tracking error         : {err_history.max():.3f} deg")
print(f"  final pointing error vs qf  : {final_pt:.3f} deg\n")

# =========================================
# Plots
# =========================================
t_nodes = np.arange(T + 1) * dt_control
w_nodes = x_value[4:7, :]        # reference omega [rad/s]

plt.figure(figsize=(7, 3))
plt.plot(t_history, q_history)
plt.plot(t_history, np.linalg.norm(q_history, axis=1))
plt.plot(t_nodes, x_value[0:4, :].T, 'o', ms=3, alpha=0.5)
plt.axvline(tf, color='k', ls=':', lw=1, alpha=0.6)
plt.legend(["q0","q1","q2","q3","|q|"])
plt.title("Quaternion  (- open-loop perturbed, o guidance)")
plt.xlabel("Time [s]"); plt.grid(); plt.show()

plt.figure(figsize=(7, 3))
plt.plot(t_history, w_history * 180/np.pi)
plt.plot(t_nodes, w_nodes.T * 180/np.pi, 'o', ms=3, alpha=0.5)
plt.axvline(tf, color='k', ls=':', lw=1, alpha=0.6)
plt.legend(["wx","wy","wz"])
plt.title("Angular Velocity [deg/s]")
plt.xlabel("Time [s]"); plt.grid(); plt.show()

plt.figure(figsize=(7, 3))
plt.plot(t_history, err_history)
plt.axvline(tf, color='k', ls=':', lw=1, alpha=0.6)
plt.title("Open-loop deviation from COGU reference  (perturbed plant)")
plt.xlabel("Time [s]"); plt.ylabel("error [deg]"); plt.grid(); plt.show()

fig, axs = plt.subplots(4, 1, sharex=True, figsize=(6, 6))
colors = ['tab:blue', 'tab:orange', 'tab:green', 'tab:red']
for i in range(4):
    axs[i].step(t_history, u_history[:, i] * 1e6, where='post', color=colors[i], label=f'VAT{i+1} (noisy)')
    axs[i].step(t_history, ucmd_history[:, i] * 1e6, where='post', color='k', lw=0.8, alpha=0.4, label='ideal')
    axs[i].set_ylabel(f'Fth{i+1} [uN]'); axs[i].grid(True); axs[i].legend(loc='upper right', fontsize=7)
axs[-1].set_xlabel('Time [s]')
fig.suptitle('Thruster force sequence [uN]  (noisy vs ideal)')
fig.tight_layout(); plt.show()
```

# Open loop (VAT, COGU) + Closed loop (RW)

```python
# =========================================
# Thruster control matrix B   (M = sum_i u_i (rth_i x fth_i))
# =========================================
B = np.column_stack([
    np.cross(rth1_double[:, 0], fth1_double[:, 0]),
    np.cross(rth2_double[:, 0], fth2_double[:, 0]),
    np.cross(rth3_double[:, 0], fth3_double[:, 0]),
    np.cross(rth4_double[:, 0], fth4_double[:, 0]),
])


# =========================================
# Quaternion helpers (Hamilton, consistent with q_dot below)
# =========================================
def quat_mult(a, b):
    a0, a1, a2, a3 = a
    b0, b1, b2, b3 = b
    return np.array([
        a0*b0 - a1*b1 - a2*b2 - a3*b3,
        a0*b1 + a1*b0 + a2*b3 - a3*b2,
        a0*b2 - a1*b3 + a2*b0 + a3*b1,
        a0*b3 + a1*b2 - a2*b1 + a3*b0,
    ])

def quat_conj(q):
    return np.array([q[0], -q[1], -q[2], -q[3]])

def quat_rotate(q, v):
    qv = np.array([0.0, *v])
    return quat_mult(quat_mult(q, qv), quat_conj(q))[1:4]


# =========================================
# Perturbations / uncertainty configuration
# =========================================
pert_seed = 0                              # RNG seed (None -> nondeterministic)
rng = np.random.default_rng(pert_seed)

# --- Thruster actuator noise (applied only when a thruster fires) ---
#   F_actual = F_nominal + thrust_offset + thrust_bias_i + N(0, thrust_sigma)
thrust_nominal     = Fth_max_double                 # nominal firing force [N]
thrust_offset      = 0.0                   # systematic mean offset on every firing [N]
thrust_sigma       = Fth_max_double/10                  # per-step Gaussian std on each firing [N]   (TUNE)
thrust_bias_sigma  = Fth_max_double/20                  # fixed thruster-to-thruster bias std [N]     (TUNE)
# fixed per-thruster bias, drawn once (thruster mismatch)
thrust_bias = thrust_bias_sigma * rng.standard_normal(4)

# --- Inertia uncertainty: true plant inertia differs from the nominal model ---
inertia_var = 0.50                         # +/- 50% multiplicative variation
def disturbance_torque(t):
    return np.array([
        1.0e-7*np.sin(2*np.pi*t/5400),
        0.8e-7*np.sin(2*np.pi*t/4200 + 1.3),
        0.6e-7*np.sin(2*np.pi*t/6800 + 2.1)
    ])

def _deficit_distribute(counts, n):
    """
    Reparte n slots [0..n-1] entre len(counts) categorías, de forma
    determinística y lo más uniforme posible, según target = counts[c].
    sum(counts) debe ser <= n (el resto queda implícito si counts no suma n).
    Devuelve una lista de longitud n con el índice de categoría por slot.

    Algoritmo: en cada paso i, se asigna el slot a la categoría con mayor
    "déficit" = target_fraccional_acumulado - ya_asignado. Generalización
    multi-categoría de Bresenham / Sainte-Laguë sequential apportionment.
    """
    n_cat = len(counts)
    assigned = [0] * n_cat
    out = [-1] * n
    for i in range(n):
        best_c, best_deficit = -1, -1.0
        for c in range(n_cat):
            if counts[c] <= 0:
                continue
            target = counts[c] * (i + 1) / n
            deficit = target - assigned[c]
            if deficit > best_deficit:
                best_deficit, best_c = deficit, c
        if best_c == -1:           # ningún target > 0 (caso todo-cero)
            best_c = n_cat - 1      # cae en categoría "ninguno" si existe
        assigned[best_c] += 1
        out[i] = best_c
    return out


def _pair_fire_pattern(a, b, n_sub, Fmax):
    """
    Para un par antagonista {A,B} (mutuamente excluyentes), devuelve un
    array de longitud n_sub con 0=ninguno, 1=A, 2=B en cada sub-slot,
    distribuido lo más uniformemente posible.
    a, b >= 0 ; a+b <= Fmax (si a+b < Fmax, algunos slots quedan en 0).
    """
    ka = int(round((a / Fmax) * n_sub))
    kb = int(round((b / Fmax) * n_sub))
    ka = min(ka, n_sub)
    kb = min(kb, n_sub - ka)          # garantiza ka+kb <= n_sub
    k_none = n_sub - ka - kb
    # orden de categorías: 0=none, 1=A, 2=B
    pattern = _deficit_distribute([k_none, ka, kb], n_sub)
    return pattern


def u_interpol_zoh(t, u_value, dt_control, n_sub=10):
    T_control = u_value.shape[1]
    index = int(np.floor(t / dt_control))
    if index >= T_control:
        return np.zeros(u_value.shape[0])
    u = u_value[:, index]
    t_local = t - index * dt_control
    dt_sub = dt_control / n_sub
    slot_idx = min(int(np.floor(t_local / dt_sub)), n_sub - 1)

    out = np.zeros(u_value.shape[0])

    # --- Par {VAT1, VAT2} ---
    pattern01 = _pair_fire_pattern(u[0], u[1], n_sub, thrust_nominal)
    cat = pattern01[slot_idx]
    if cat == 1:
        out[0] = thrust_nominal
    elif cat == 2:
        out[1] = thrust_nominal

    # --- Par {VAT3, VAT4} ---
    pattern23 = _pair_fire_pattern(u[2], u[3], n_sub, thrust_nominal)
    cat = pattern23[slot_idx]
    if cat == 1:
        out[2] = thrust_nominal
    elif cat == 2:
        out[3] = thrust_nominal

    return out


def apply_thruster_noise(u_cmd):
    """Inject Gaussian firing noise + offset + fixed bias on firing thrusters."""
    fired = u_cmd > 0.0
    noisy = u_cmd + thrust_offset + thrust_bias + thrust_sigma * rng.standard_normal(4)
    # only firing thrusters are perturbed; force stays non-negative
    return np.where(fired, np.clip(noisy, 0.0, None), 0.0)


# =========================================
# Reference: guidance interpolation, then eternal fixed setpoint
# =========================================
# Setpoint for eternal tracking (default = final guidance attitude, zero rate).
q_set = x_value[0:4, T].copy()
q_set /= np.linalg.norm(q_set)
w_set = np.zeros(3)

def reference(t):
    # --- eternal tracking phase: fixed setpoint ---
    if t >= tf:
        return q_set, w_set

    # --- guidance phase: linear interpolation between nodes ---
    idx = int(np.floor(t / dt_control))
    if idx >= T:
        return q_set, w_set
    alpha = (t - idx * dt_control) / dt_control
    q0 = x_value[0:4, idx]
    q1 = x_value[0:4, idx + 1]
    if np.dot(q0, q1) < 0.0:               # shortest path
        q1 = -q1
    q_ref = (1 - alpha) * q0 + alpha * q1
    q_ref /= np.linalg.norm(q_ref)
    w_ref = (1 - alpha) * x_value[4:7, idx] + alpha * x_value[4:7, idx + 1]
    return q_ref, w_ref


# =========================================
# Reaction wheel configuration
#   3 orthogonal wheels aligned with body x, y, z axes
#   Aw : 3 x 3  ->  spin axes = body axes
# =========================================
Aw = np.eye(3)                        # wheel i -> body axis i
Nw = Aw.shape[1]                      # = 3
Aw_pinv = np.eye(3)                   # trivial allocation: tau_w_cmd = -tau_pd per axis

Jw         = 4.77e-6                  # wheel axial inertia [kg*m^2]        (TUNE)
tau_w_max  = 1.0e-3                   # max motor torque per wheel [N*m]    (TUNE)
Omega_max  = 8000.0 * (2*np.pi/60)    # wheel speed saturation [rad/s]  (= 8000 rpm)
tau_motor  = 0.10                     # torque-servo time constant [s]      (TUNE)


# =========================================
# Servo controller  (runs ONLY at the control rate; output is ZOH-held)
#   Outer loop : attitude PD       -> desired body torque tau_pd
#   Allocation : tau_w_cmd = -Aw^+ tau_pd   (body receives -Aw*tau_w)
# =========================================
def servo_control(t, x, params):
    q = x[0:4]; w = x[4:7]
    q_ref, w_ref = params["ref"](t)

    q_err = quat_mult(quat_conj(q_ref), q)
    if q_err[0] < 0.0:                       # avoid unwinding (shortest path)
        q_err = -q_err

    att_err    = q_err[1:4]                  # ~ theta/2 for small errors
    w_ref_body = quat_rotate(quat_conj(q_err), w_ref)
    w_err      = w - w_ref_body

    tau_pd    = -(params["Kp"] @ att_err) - (params["Kd"] @ w_err)
    tau_w_cmd = -params["Aw_pinv"] @ tau_pd
    tau_w_cmd = np.clip(tau_w_cmd, -params["tau_w_max"], params["tau_w_max"])
    return tau_pd, tau_w_cmd


# =========================================
# Dynamics  (open-loop thrusters + wheels; tau_w_cmd is a held ZOH input)
#   x = [q(4), w(3), Omega(Nw), tau_w(Nw)]
#   NOTE: params["I"] is the TRUE (perturbed) plant inertia.
# =========================================
def dynamics(t, x, u, tau_w_cmd, params):
    q     = x[0:4]
    w     = x[4:7]
    Omega = x[7:7+Nw]
    tau_w = x[7+Nw:7+2*Nw]                    # actual motor torque (servo state)

    I, I_inv = params["I"], params["I_inv"]
    Aw, Jw   = params["Aw"], params["Jw"]

    # total angular momentum (body + wheels)
    h_w = Aw @ (Jw * Omega)
    L   = I @ w + h_w

    # external torque: open-loop thrusters
    tau_ff = (params["B"] @ u) if params["use_thrusters"] else np.zeros(3)

    # wheel reaction on the body
    tau_rw = -Aw @ tau_w
    tau_d = disturbance_torque(t)
    # Euler with wheel gyroscopic term included
    w_dot = I_inv @ (tau_ff + tau_rw - np.cross(w, L)+tau_d)

    # wheel speed dynamics:  Jw * Omega_dot = tau_w
    Omega_dot = tau_w / Jw

    # inner torque servo: motor torque chases the HELD command (continuous actuator dyn.)
    tau_w_dot = (tau_w_cmd - tau_w) / params["tau_motor"]

    # quaternion kinematics
    wx, wy, wz = w
    q0, q1, q2, q3 = q
    q_dot = 0.5 * np.array([
        -wx*q1 - wy*q2 - wz*q3,
         wx*q0 + wz*q2 - wy*q3,
         wy*q0 - wz*q1 + wx*q3,
         wz*q0 + wy*q1 - wx*q2
    ])

    return np.concatenate((q_dot, w_dot, Omega_dot, tau_w_dot))


# =========================================
# RK4 integrator  (u and tau_w_cmd held constant across the step)
# =========================================
def rk4_step(f, t, x, u, tau_w_cmd, dt, params):
    k1 = f(t,        x,             u, tau_w_cmd, params)
    k2 = f(t + dt/2, x + dt/2 * k1, u, tau_w_cmd, params)
    k3 = f(t + dt/2, x + dt/2 * k2, u, tau_w_cmd, params)
    k4 = f(t + dt,   x + dt   * k3, u, tau_w_cmd, params)
    return x + dt/6 * (k1 + 2*k2 + 2*k3 + k4)


# =========================================
# Setup
# =========================================
I_nominal = Jsat_double.copy()             # controller's model (assumed known)

# --- Inertia uncertainty: build the TRUE plant inertia ---
#   Multiplicative +/- inertia_var perturbation, kept symmetric and
#   re-drawn until positive-definite (a satellite inertia stays PD at 50%).
while True:
    P = 1.0 + inertia_var * (2.0 * rng.random((3, 3)) - 1.0)   # entries in [1-v, 1+v]
    P = 0.5 * (P + P.T)                                        # symmetric
    I_true = I_nominal * P                                     # element-wise
    if np.all(np.linalg.eigvalsh(I_true) > 0.0):
        break

print("Nominal inertia diag [kg.m^2]:", np.diag(I_nominal))
print("True    inertia diag [kg.m^2]:", np.diag(I_true))
# per-axis relative error on the diagonal (safe; diagonal is always nonzero)
diag_relerr = 100.0 * (np.diag(I_true) - np.diag(I_nominal)) / np.diag(I_nominal)
print("Diagonal inertia error (%):", np.round(diag_relerr, 1))
print("True inertia eigenvalues:", np.linalg.eigvalsh(I_true))

wn   = 0.3                          # rad/s, tracking bandwidth (TUNE)
zeta = 1.0
Kp = 0.2 * wn**2 * I_nominal       # gains designed on NOMINAL inertia (controller blind to truth)
Kd = 2.0 * zeta * wn * I_nominal

params = {
    "I":             I_true,       # <-- TRUE plant inertia
    "I_inv":         np.linalg.inv(I_true),
    "B":             B,
    "Kp":            Kp,
    "Kd":            Kd,
    "ref":           reference,
    "Aw":            Aw,
    "Aw_pinv":       Aw_pinv,
    "Jw":            Jw,
    "tau_w_max":     tau_w_max,
    "tau_motor":     tau_motor,
    "use_thrusters": True,         # False -> wheels only
}

# Initial state: guidance node 0 + wheels at rest + zero motor torque
x = np.concatenate((x_value[:, 0].copy(), np.zeros(Nw), np.zeros(Nw)))

dt_state   = 0.1                          # integration step [s]            (10 Hz, RK4)
dt_ctrl    = 1.0                          # WHEEL controller period [s]     (1 Hz, closed loop)
dt_control = tau_double                   # thruster guidance step (ZOH) [s]
hold_time  = 120.0                        # eternal-tracking time after guidance [s] (TUNE)
sim_time   = tf + hold_time
N = int(round(sim_time / dt_state)) + 1

control_every = int(round(dt_ctrl / dt_state))   # integration steps per control tick (= 10)
if abs(dt_ctrl / dt_state - control_every) > 1e-9:
    print(f"[warning] dt_ctrl={dt_ctrl}s is not a multiple of dt_state={dt_state}s")
if abs(dt_control / dt_state - round(dt_control / dt_state)) > 1e-9:
    print(f"[warning] dt_state={dt_state}s does not divide thruster dt_control={dt_control}s")

q_history       = np.zeros((N, 4))
w_history       = np.zeros((N, 3))
u_history       = np.zeros((N, 4))     # NOISY thruster force actually applied
ucmd_history    = np.zeros((N, 4))     # IDEAL on/off command (for comparison)
Omega_history   = np.zeros((N, Nw))
tauw_history    = np.zeros((N, Nw))     # actual motor torque (servo state)
tauwcmd_history = np.zeros((N, Nw))     # held ZOH command (1 Hz staircase)
taurw_history   = np.zeros((N, 3))      # control torque delivered to the body by the wheels
hw_history      = np.zeros((N, 3))      # wheel angular momentum in body frame
t_history       = np.zeros(N)

t = 0.0
sat_warned = False
tau_w_cmd = np.zeros(Nw)                # last held wheel command (ZOH)

# =========================================
# Simulation loop
#   Thrusters : open-loop ZOH at guidance step  (+ actuator noise)
#   Wheels    : closed-loop servo sampled @ 1 Hz, ZOH-held
#   Plant     : RK4 @ 0.1 s  with TRUE (perturbed) inertia
# =========================================
for i in range(N):
    q_history[i]     = x[0:4]
    w_history[i]     = x[4:7]
    Omega_history[i] = x[7:7+Nw]
    tauw_history[i]  = x[7+Nw:7+2*Nw]
    taurw_history[i] = -Aw @ x[7+Nw:7+2*Nw]
    hw_history[i]    = Aw @ (Jw * x[7:7+Nw])
    t_history[i]     = t

    # --- thrusters: ideal ZOH command, then inject actuator noise ---
    u_cmd          = u_interpol_zoh(t, u_value, dt_control)   # ideal 0 / 50uN
    u_current      = apply_thruster_noise(u_cmd)              # noisy applied force
    ucmd_history[i] = u_cmd
    u_history[i]    = u_current

    # --- wheel controller: fires only every dt_ctrl seconds, then ZOH-held ---
    if i % control_every == 0:
        _, tau_w_cmd = servo_control(t, x, params)
    tauwcmd_history[i] = tau_w_cmd

    x = rk4_step(dynamics, t, x, u_current, tau_w_cmd, dt_state, params)
    x[0:4] /= np.linalg.norm(x[0:4])          # renormalize q

    # wheel speed saturation
    Omega = x[7:7+Nw]
    if np.any(np.abs(Omega) > Omega_max):
        if not sat_warned:
            print("[warning] wheel speed saturated -> momentum dumping (desaturation) needed")
            sat_warned = True
        x[7:7+Nw] = np.clip(Omega, -Omega_max, Omega_max)

    t += dt_state


# =========================================
# Plots
# =========================================
t_nodes = np.arange(T + 1) * dt_control

plt.figure(figsize=(7, 3))
plt.plot(t_history, q_history)
plt.plot(t_history, np.linalg.norm(q_history, axis=1))
plt.plot(t_nodes, x_value[0:4, :].T, 'o', ms=3, alpha=0.5)
plt.axvline(tf, color='k', ls=':', lw=1, alpha=0.6)
plt.legend(["q0","q1","q2","q3","|q|"])
plt.title("Quaternion  (VAT @1Hz, o guidance)")
plt.xlabel("Time [s]"); plt.grid(); plt.show()

plt.figure(figsize=(7, 3))
plt.plot(t_history, w_history * 180/np.pi)
plt.plot(t_nodes, x_value[4:7, :].T * 180/np.pi, 'o', ms=3, alpha=0.5)
plt.axvline(tf, color='k', ls=':', lw=1, alpha=0.6)
plt.legend(["wx","wy","wz"])
plt.title("Angular Velocity [deg/s]")
plt.xlabel("Time [s]"); plt.grid(); plt.show()

fig, axs = plt.subplots(4, 1, sharex=True, figsize=(6, 6))

colors = ['tab:blue', 'tab:orange', 'tab:green', 'tab:red']

for i in range(4):
    axs[i].step(t_history, u_history[:, i] * 1e6, where='post',
                color=colors[i], label=f'VAT{i+1} (noisy)')
    axs[i].step(t_history, ucmd_history[:, i] * 1e6, where='post',
                color='k', lw=0.8, alpha=0.4, label='ideal')
    axs[i].set_ylabel(f'Fth{i+1} [uN]')
    axs[i].grid(True)
    axs[i].legend(loc='upper right', fontsize=7)
    axs[i].tick_params(labelsize=10)

axs[-1].set_xlabel('Time [s]')
fig.suptitle('Thruster force sequence [uN]  (noisy vs ideal 25uN)')
fig.tight_layout()
plt.show()

plt.figure(figsize=(7, 3))
plt.plot(t_history, Omega_history * 60/(2*np.pi))
plt.axhline( Omega_max * 60/(2*np.pi), color='r', ls=':', lw=2, alpha=0.5)
plt.axhline(-Omega_max * 60/(2*np.pi), color='r', ls=':', lw=2, alpha=0.5)
plt.axvline(tf, color='k', ls=':', lw=1, alpha=0.6)
plt.legend([f"RW{i+1}" for i in range(Nw)])
plt.title("Wheel speed [rpm]  (red = saturation)")
plt.xlabel("Time [s]"); plt.grid(); plt.show()

plt.figure(figsize=(7, 3))
for i in range(Nw):
    plt.step(t_history, tauwcmd_history[:, i] * 1e3, where='post', label=f"cmd RW{i+1}")
plt.plot(t_history, tauw_history * 1e3, '--', alpha=0.7)
plt.axvline(tf, color='k', ls=':', lw=1, alpha=0.6)
plt.legend()
plt.title("Wheel torque [mN.m]  (-- actual servo,  step = 1 Hz command)")
plt.xlabel("Time [s]"); plt.grid(); plt.show()

plt.figure(figsize=(7, 3))
plt.plot(t_history, taurw_history * 1e6)
plt.axvline(tf, color='k', ls=':', lw=1, alpha=0.6)
plt.legend(["tau_x","tau_y","tau_z"])
plt.title("Control torque on body from wheels [uN.m]")
plt.xlabel("Time [s]"); plt.grid(); plt.show()

plt.figure(figsize=(7, 3))
plt.plot(t_history, hw_history * 1e3)
plt.plot(t_history, np.linalg.norm(hw_history, axis=1) * 1e3, 'k--')
plt.axvline(tf, color='k', ls=':', lw=1, alpha=0.6)
plt.legend(["hx","hy","hz","|h|"])
plt.title("Wheel angular momentum [mN.m.s]")
plt.xlabel("Time [s]"); plt.grid(); plt.show()
```

# Closed loop (VAT, COGU)

```python
# ============================================================================
#  VAT OPEN-LOOP COGU on a PERTURBED PLANT  (no RW)
# ----------------------------------------------------------------------------
#  Reaction wheels removed. The COGU pulse train u_value is applied open-loop
#  to the TRUE perturbed plant. State = [q(4), omega(3)]  (x_value[4:7] is omega).
#  Perturbations:
#     - inertia error (I_true vs I_nom)
#     - actuator noise (offset + per-thruster bias + Gaussian, only while firing)
#     - tau_dist : external disturbance torque, ALWAYS ON (even with VAT off)
# ============================================================================

import numpy as np
import matplotlib.pyplot as plt

# =========================================
# Thruster control matrix B
# =========================================
B = np.column_stack([
    np.cross(rth1_double[:, 0], fth1_double[:, 0]),
    np.cross(rth2_double[:, 0], fth2_double[:, 0]),
    np.cross(rth3_double[:, 0], fth3_double[:, 0]),
    np.cross(rth4_double[:, 0], fth4_double[:, 0]),
])

I_nominal = Jsat_double.copy()

# =========================================
# Perturbations / uncertainty
# =========================================
pert_seed = 0
rng = np.random.default_rng(pert_seed)

# --- actuator noise (only while a thruster fires) ---
thrust_nominal     = Fth_max_double
thrust_offset      = 0.0                    # systematic offset on every firing [N]   (TUNE)
thrust_sigma       = Fth_max_double / 10    # per-firing Gaussian std [N]
thrust_bias_sigma  = Fth_max_double / 20    # fixed thruster-to-thruster bias std [N]
thrust_bias        = thrust_bias_sigma * rng.standard_normal(4)

# --- inertia uncertainty ---
inertia_var = 0.05                        # +/- 5% (set 0.0 to disable)             (TUNE)
# --- EXTERNAL disturbance torque: ALWAYS applied, even when VAT = 0 ---
def disturbance_torque(t):
    return np.array([
        1.0e-7*np.sin(2*np.pi*t/5400),
        0.8e-7*np.sin(2*np.pi*t/4200 + 1.3),
        0.6e-7*np.sin(2*np.pi*t/6800 + 2.1)
    ])

def _deficit_distribute(counts, n):
    """
    Reparte n slots [0..n-1] entre len(counts) categorías, de forma
    determinística y lo más uniforme posible, según target = counts[c].
    sum(counts) debe ser <= n (el resto queda implícito si counts no suma n).
    Devuelve una lista de longitud n con el índice de categoría por slot.

    Algoritmo: en cada paso i, se asigna el slot a la categoría con mayor
    "déficit" = target_fraccional_acumulado - ya_asignado. Generalización
    multi-categoría de Bresenham / Sainte-Laguë sequential apportionment.
    """
    n_cat = len(counts)
    assigned = [0] * n_cat
    out = [-1] * n
    for i in range(n):
        best_c, best_deficit = -1, -1.0
        for c in range(n_cat):
            if counts[c] <= 0:
                continue
            target = counts[c] * (i + 1) / n
            deficit = target - assigned[c]
            if deficit > best_deficit:
                best_deficit, best_c = deficit, c
        if best_c == -1:           # ningún target > 0 (caso todo-cero)
            best_c = n_cat - 1      # cae en categoría "ninguno" si existe
        assigned[best_c] += 1
        out[i] = best_c
    return out


def _pair_fire_pattern(a, b, n_sub, Fmax):
    """
    Para un par antagonista {A,B} (mutuamente excluyentes), devuelve un
    array de longitud n_sub con 0=ninguno, 1=A, 2=B en cada sub-slot,
    distribuido lo más uniformemente posible.
    a, b >= 0 ; a+b <= Fmax (si a+b < Fmax, algunos slots quedan en 0).
    """
    ka = int(round((a / Fmax) * n_sub))
    kb = int(round((b / Fmax) * n_sub))
    ka = min(ka, n_sub)
    kb = min(kb, n_sub - ka)          # garantiza ka+kb <= n_sub
    k_none = n_sub - ka - kb
    # orden de categorías: 0=none, 1=A, 2=B
    pattern = _deficit_distribute([k_none, ka, kb], n_sub)
    return pattern


def u_interpol_zoh(t, u_value, dt_control, n_sub=10):
    T_control = u_value.shape[1]
    index = int(np.floor(t / dt_control))
    if index >= T_control:
        return np.zeros(u_value.shape[0])
    u = u_value[:, index]
    t_local = t - index * dt_control
    dt_sub = dt_control / n_sub
    slot_idx = min(int(np.floor(t_local / dt_sub)), n_sub - 1)

    out = np.zeros(u_value.shape[0])

    # --- Par {VAT1, VAT2} ---
    pattern01 = _pair_fire_pattern(u[0], u[1], n_sub, thrust_nominal)
    cat = pattern01[slot_idx]
    if cat == 1:
        out[0] = thrust_nominal
    elif cat == 2:
        out[1] = thrust_nominal

    # --- Par {VAT3, VAT4} ---
    pattern23 = _pair_fire_pattern(u[2], u[3], n_sub, thrust_nominal)
    cat = pattern23[slot_idx]
    if cat == 1:
        out[2] = thrust_nominal
    elif cat == 2:
        out[3] = thrust_nominal

    return out

def apply_thruster_noise(u_cmd):
    fired = u_cmd > 0.0
    noisy = u_cmd + thrust_offset + thrust_bias + thrust_sigma * rng.standard_normal(4)
    return np.where(fired, np.clip(noisy, 0.0, None), 0.0)

# =========================================
# Dynamics  (state = [q, omega])  with disturbance torque, no wheels
# =========================================
def dynamics(t, x, u, params):
    q = x[0:4]; w = x[4:7]
    I, I_inv = params["I"], params["I_inv"]
    tau_ff = (params["B"] @ u) if params["use_thrusters"] else np.zeros(3)
    tau_d = disturbance_torque(t)
    w_dot = I_inv @ (tau_ff + tau_d - np.cross(w, I @ w))   # + disturbance
    wx, wy, wz = w; q0, q1, q2, q3 = q
    q_dot = 0.5 * np.array([
        -wx*q1 - wy*q2 - wz*q3,
         wx*q0 + wz*q2 - wy*q3,
         wy*q0 - wz*q1 + wx*q3,
         wz*q0 + wy*q1 - wx*q2])
    return np.concatenate((q_dot, w_dot))

def rk4_step(f, t, x, u, dt, params):
    k1 = f(t,        x,             u, params)
    k2 = f(t + dt/2, x + dt/2 * k1, u, params)
    k3 = f(t + dt/2, x + dt/2 * k2, u, params)
    k4 = f(t + dt,   x + dt   * k3, u, params)
    return x + dt/6 * (k1 + 2*k2 + 2*k3 + k4)

# =========================================
# CLOSED-LOOP VAT CORRECTOR
# =========================================

def quat_conj(q):
    return np.array([q[0], -q[1], -q[2], -q[3]])

def quat_mult(q1, q2):

    w1,x1,y1,z1 = q1
    w2,x2,y2,z2 = q2

    return np.array([
        w1*w2 - x1*x2 - y1*y2 - z1*z2,
        w1*x2 + x1*w2 + y1*z2 - z1*y2,
        w1*y2 - x1*z2 + y1*w2 + z1*x2,
        w1*z2 + x1*y2 - y1*x2 + z1*w2
    ])

F_max = thrust_nominal

def allocate_vat(tau_des):

    duty = np.zeros(4)

    for k in range(4):

        di = B[:,k]

        proj = np.dot(tau_des,di)/(np.dot(di,di)+1e-30)

        duty[k] = proj/F_max

    duty = np.clip(duty,0.0,1.0)

    if duty[0] >= duty[1]:
        duty[1] = 0.0
    else:
        duty[0] = 0.0

    if duty[2] >= duty[3]:
        duty[3] = 0.0
    else:
        duty[2] = 0.0

    return duty


Km   = 5.0
Tm   = 0.50

Uon  = 0.45
Uoff = 0.15

def pwpf_update(f,fire,c,dt):

    e = c-fire

    f = f + dt*(Km*e-f)/Tm

    new = fire.copy()

    new[f>=Uon]  = 1.0
    new[f<=Uoff] = 0.0

    return f,new

# =========================================
# Setup
# =========================================
while True:                                  # TRUE perturbed inertia
    P = 1.0 + inertia_var * (2.0 * rng.random((3, 3)) - 1.0)
    P = 0.5 * (P + P.T)
    I_true = I_nominal * P
    if np.all(np.linalg.eigvalsh(I_true) > 0.0):
        break
print("Nominal inertia diag [kg.m^2]:", np.diag(I_nominal))
print("True    inertia diag [kg.m^2]:", np.diag(I_true))
print("Diagonal inertia error (%):", np.round(100*(np.diag(I_true)-np.diag(I_nominal))/np.diag(I_nominal), 1))

params = {
    "I": I_true, "I_inv": np.linalg.inv(I_true), "B": B,
    "use_thrusters": True,
}

# =========================================
# Feedback controller sizing
# =========================================

tau_axis   = F_max*np.max(np.abs(B),axis=1)
alpha_axis = tau_axis/np.diag(I_nominal)

alpha_max = np.max(alpha_axis)

brake_angle = np.deg2rad(20.0)
w_lim = np.sqrt(2.0*alpha_max*brake_angle)
K_att = w_lim/np.sin(brake_angle/2.0)
Kd_fb = 2e-3

f_pwpf = np.zeros(4)
fire   = np.zeros(4)

# Initial state: [q0, omega0]
x = x_value[:, 0].copy()
x[0:4] /= np.linalg.norm(x[0:4])

dt_state   = 0.1
dt_control = tau_double
hold_time  = 600.0
sim_time   = tf + hold_time
N = int(round(sim_time / dt_state)) + 1

q_history = np.zeros((N, 4)); w_history = np.zeros((N, 3))
u_history = np.zeros((N, 4)); ucmd_history = np.zeros((N, 4))
err_history = np.zeros(N); t_history = np.zeros(N)

def q_ref_at(t):
    idx = min(int(np.floor(t / dt_control)), T - 1)
    a = (t - idx * dt_control) / dt_control
    q0 = x_value[0:4, idx]; q1 = x_value[0:4, idx + 1]
    if np.dot(q0, q1) < 0.0: q1 = -q1
    qr = (1 - a) * q0 + a * q1
    return qr / np.linalg.norm(qr)

q_end = x_value[0:4, T].copy()
q_end /= np.linalg.norm(q_end)

t = 0.0
for i in range(N):
    q_history[i] = x[0:4]; w_history[i] = x[4:7]; t_history[i] = t

    if t <= tf:
        qr = q_ref_at(t)
    else:
        qr = q_end
    err_history[i] = 2.0 * np.degrees(np.arccos(np.clip(abs(np.dot(x[0:4], qr)), 0, 1)))

    if t <= tf:

        # ===============================
        # OPEN LOOP COGU
        # ===============================

        u_cmd = u_interpol_zoh(
            t,
            u_value,
            dt_control
        )

    else:

        # ===============================
        # CLOSED LOOP CORRECTION
        # ===============================

        q = x[0:4]
        w = x[4:7]

        q_err = quat_mult(
            quat_conj(q_end),
            q
        )

        if q_err[0] < 0.0:
            q_err = -q_err

        att_err = q_err[1:4]

        w_cmd = -K_att*att_err

        nrm = np.linalg.norm(w_cmd)

        if nrm > w_lim:
            w_cmd *= w_lim/nrm

        tau_des = -Kd_fb*(w-w_cmd)

        duty = allocate_vat(tau_des)

        f_pwpf, fire = pwpf_update(
            f_pwpf,
            fire,
            duty,
            dt_state
        )

        u_cmd = fire*F_max

    u_current = apply_thruster_noise(u_cmd)
    ucmd_history[i] = u_cmd; u_history[i] = u_current

    x = rk4_step(dynamics, t, x, u_current, dt_state, params)
    x[0:4] /= np.linalg.norm(x[0:4])
    t += dt_state

q_end = x_value[0:4, T] / np.linalg.norm(x_value[0:4, T])
final_pt = 2.0 * np.degrees(np.arccos(np.clip(abs(np.dot(q_history[-1], q_end)), 0, 1)))
print(f"\nOpen-loop drift vs COGU reference:")
print(f"  final tracking error       : {err_history[-1]:.3f} deg")
print(f"  max tracking error         : {err_history.max():.3f} deg")
print(f"  final pointing error vs qf  : {final_pt:.3f} deg\n")

# =========================================
# Plots
# =========================================
t_nodes = np.arange(T + 1) * dt_control
w_nodes = x_value[4:7, :]        # reference omega [rad/s]

plt.figure(figsize=(7, 3))
plt.plot(t_history, q_history)
plt.plot(t_history, np.linalg.norm(q_history, axis=1))
plt.plot(t_nodes, x_value[0:4, :].T, 'o', ms=3, alpha=0.5)
plt.axvline(tf, color='k', ls=':', lw=1, alpha=0.6)
plt.legend(["q0","q1","q2","q3","|q|"])
plt.title("Quaternion  (- open-loop perturbed, o guidance)")
plt.xlabel("Time [s]"); plt.grid(); plt.show()

plt.figure(figsize=(7, 3))
plt.plot(t_history, w_history * 180/np.pi)
plt.plot(t_nodes, w_nodes.T * 180/np.pi, 'o', ms=3, alpha=0.5)
plt.axvline(tf, color='k', ls=':', lw=1, alpha=0.6)
plt.legend(["wx","wy","wz"])
plt.title("Angular Velocity [deg/s]")
plt.xlabel("Time [s]"); plt.grid(); plt.show()

plt.figure(figsize=(7, 3))
plt.plot(t_history, err_history)
plt.axvline(tf, color='k', ls=':', lw=1, alpha=0.6)
plt.title("Open-loop deviation from COGU reference  (perturbed plant)")
plt.xlabel("Time [s]"); plt.ylabel("error [deg]"); plt.grid(); plt.show()

fig, axs = plt.subplots(4, 1, sharex=True, figsize=(6, 6))
colors = ['tab:blue', 'tab:orange', 'tab:green', 'tab:red']
for i in range(4):
    axs[i].step(t_history, u_history[:, i] * 1e6, where='post', color=colors[i], label=f'VAT{i+1} (noisy)')
    axs[i].step(t_history, ucmd_history[:, i] * 1e6, where='post', color='k', lw=0.8, alpha=0.4, label='ideal')
    axs[i].set_ylabel(f'Fth{i+1} [uN]'); axs[i].grid(True); axs[i].legend(loc='upper right', fontsize=7)
axs[-1].set_xlabel('Time [s]')
fig.suptitle('Thruster force sequence [uN]  (noisy vs ideal)')
fig.tight_layout(); plt.show()
```

# Closed loop (RW only)

```python
import numpy as np
import matplotlib.pyplot as plt

# =========================================
# Quaternion helpers (Hamilton, consistent with q_dot below)
# =========================================
def quat_mult(a, b):
    a0, a1, a2, a3 = a
    b0, b1, b2, b3 = b
    return np.array([
        a0*b0 - a1*b1 - a2*b2 - a3*b3,
        a0*b1 + a1*b0 + a2*b3 - a3*b2,
        a0*b2 - a1*b3 + a2*b0 + a3*b1,
        a0*b3 + a1*b2 - a2*b1 + a3*b0,
    ])

def quat_conj(q):
    return np.array([q[0], -q[1], -q[2], -q[3]])

def quat_rotate(q, v):
    qv = np.array([0.0, *v])
    return quat_mult(quat_mult(q, qv), quat_conj(q))[1:4]


# =========================================
# Fixed quaternion setpoint  (the ONLY reference: hold this attitude forever)
# =========================================
q_set = x_value[0:4, T].copy()
q_set /= np.linalg.norm(q_set)
w_set = np.zeros(3)                         # zero rate at the setpoint

def reference(t):
    return q_set, w_set                     # constant setpoint (no guidance)


# =========================================
# Reaction wheel configuration
#   3 orthogonal wheels aligned with body x, y, z axes
#   Aw : 3 x 3  ->  spin axes = body axes
# =========================================
Aw = np.eye(3)                        # wheel i -> body axis i
Nw = Aw.shape[1]                      # = 3
Aw_pinv = np.eye(3)                   # trivial allocation: tau_w_cmd = -tau_pd per axis

Jw         = 4.77e-6                   # wheel axial inertia [kg*m^2]        (TUNE)
tau_w_max  = 1.0e-3                   # max motor torque per wheel [N*m]    (TUNE)
Omega_max  = 8000.0*(2*np.pi/60)                    # wheel speed saturation [rad/s]
tau_motor  = 0.10                     # torque-servo time constant [s]      (TUNE)


# =========================================
# Servo controller  (runs ONLY at the control rate; output is ZOH-held)
#   Outer loop : attitude PD       -> desired body torque tau_pd
#   Allocation : tau_w_cmd = -Aw^+ tau_pd   (body receives -Aw*tau_w)
# =========================================
def servo_control(t, x, params):
    q = x[0:4]; w = x[4:7]
    q_ref, w_ref = params["ref"](t)

    q_err = quat_mult(quat_conj(q_ref), q)
    if q_err[0] < 0.0:                       # avoid unwinding (shortest path)
        q_err = -q_err

    att_err    = q_err[1:4]                  # ~ theta/2 for small errors
    w_ref_body = quat_rotate(quat_conj(q_err), w_ref)
    w_err      = w - w_ref_body

    tau_pd    = -(params["Kp"] @ att_err) - (params["Kd"] @ w_err)
    tau_w_cmd = -params["Aw_pinv"] @ tau_pd
    tau_w_cmd = np.clip(tau_w_cmd, -params["tau_w_max"], params["tau_w_max"])
    return tau_pd, tau_w_cmd


# =========================================
# Dynamics  (reaction wheels only; tau_w_cmd is a held ZOH input)
#   x = [q(4), w(3), Omega(Nw), tau_w(Nw)]
# =========================================
def dynamics(t, x, tau_w_cmd, params):
    q     = x[0:4]
    w     = x[4:7]
    Omega = x[7:7+Nw]
    tau_w = x[7+Nw:7+2*Nw]                    # actual motor torque (servo state)

    I, I_inv = params["I"], params["I_inv"]
    Aw, Jw   = params["Aw"], params["Jw"]

    # total angular momentum (body + wheels)
    h_w = Aw @ (Jw * Omega)
    tau_d = disturbance_torque(t)
    L   = I @ w + h_w + tau_d

    # wheel reaction on the body (no external torque -> wheels only)
    tau_rw = -Aw @ tau_w
    w_dot  = I_inv @ (tau_rw - np.cross(w, L))   # Euler + wheel gyroscopic term

    # wheel speed dynamics:  Jw * Omega_dot = tau_w
    Omega_dot = tau_w / Jw

    # inner torque servo: motor torque chases the HELD command (continuous actuator dyn.)
    tau_w_dot = (tau_w_cmd - tau_w) / params["tau_motor"]

    # quaternion kinematics
    wx, wy, wz = w
    q0, q1, q2, q3 = q
    q_dot = 0.5 * np.array([
        -wx*q1 - wy*q2 - wz*q3,
         wx*q0 + wz*q2 - wy*q3,
         wy*q0 - wz*q1 + wx*q3,
         wz*q0 + wy*q1 - wx*q2
    ])

    return np.concatenate((q_dot, w_dot, Omega_dot, tau_w_dot))


# =========================================
# RK4 integrator  (tau_w_cmd held constant across the step)
# =========================================
def rk4_step(f, t, x, tau_w_cmd, dt, params):
    k1 = f(t,        x,             tau_w_cmd, params)
    k2 = f(t + dt/2, x + dt/2 * k1, tau_w_cmd, params)
    k3 = f(t + dt/2, x + dt/2 * k2, tau_w_cmd, params)
    k4 = f(t + dt,   x + dt   * k3, tau_w_cmd, params)
    return x + dt/6 * (k1 + 2*k2 + 2*k3 + k4)


# =========================================
# Setup
# =========================================
I = I_true.copy()

wn   = 0.3                          # rad/s, tracking bandwidth (TUNE)
zeta = 1.0
Kp = 0.2 * wn**2 * I               # att_err ~ theta/2  -> factor 2
Kd = 2.0 * zeta * wn * I

params = {
    "I":          I,
    "I_inv":      np.linalg.inv(I),
    "Kp":         Kp,
    "Kd":         Kd,
    "ref":        reference,
    "Aw":         Aw,
    "Aw_pinv":    Aw_pinv,
    "Jw":         Jw,
    "tau_w_max":  tau_w_max,
    "tau_motor":  tau_motor,
}

# --- Initial conditions (TUNE) ---
q0_init = np.array([0.0, 0.0, 0.0, 1.0])   # initial attitude (here: 180 deg off setpoint)
q0_init /= np.linalg.norm(q0_init)
w0_init = np.zeros(3)                       # initial body rate [rad/s]

# Full initial state: attitude + rate + wheels at rest + zero motor torque
x = np.concatenate((q0_init, w0_init, np.zeros(Nw), np.zeros(Nw)))

dt_state   = 0.1                           # integration step [s]            (10 Hz)
dt_control = 1.0                           # CONTROLLER update period [s]    (reacts at 1 Hz)
sim_time   = 1300.0                         # total simulation time [s] (TUNE)
N = int(round(sim_time / dt_state)) + 1

control_every = int(round(dt_control / dt_state))   # integration steps per control tick (= 10)
if abs(dt_control / dt_state - control_every) > 1e-9:
    print(f"[warning] dt_control={dt_control}s is not a multiple of dt_state={dt_state}s")

q_history       = np.zeros((N, 4))
w_history       = np.zeros((N, 3))
Omega_history   = np.zeros((N, Nw))
tauw_history    = np.zeros((N, Nw))     # actual motor torque (servo state)
tauwcmd_history = np.zeros((N, Nw))     # held ZOH command (1 Hz staircase)
taurw_history   = np.zeros((N, 3))      # control torque delivered to the body by the wheels
hw_history      = np.zeros((N, 3))      # wheel angular momentum in body frame
qerr_history    = np.zeros((N, 3))      # attitude error vector part
t_history       = np.zeros(N)

t = 0.0
sat_warned = False
tau_w_cmd = np.zeros(Nw)                # last held command (ZOH)

# =========================================
# Simulation loop  (controller at 1 Hz, plant integrated at 10 Hz)
# =========================================
for i in range(N):
    q_history[i]     = x[0:4]
    w_history[i]     = x[4:7]
    Omega_history[i] = x[7:7+Nw]
    tauw_history[i]  = x[7+Nw:7+2*Nw]
    taurw_history[i] = -Aw @ x[7+Nw:7+2*Nw]
    hw_history[i]    = Aw @ (Jw * x[7:7+Nw])
    t_history[i]     = t

    # attitude error (vector part of delta-q vs setpoint), for diagnostics
    q_err = quat_mult(quat_conj(q_set), x[0:4])
    if q_err[0] < 0.0:
        q_err = -q_err
    qerr_history[i] = q_err[1:4]

    # --- controller fires only every dt_control seconds, then ZOH-held ---
    if i % control_every == 0:
        _, tau_w_cmd = servo_control(t, x, params)
    tauwcmd_history[i] = tau_w_cmd

    x = rk4_step(dynamics, t, x, tau_w_cmd, dt_state, params)
    x[0:4] /= np.linalg.norm(x[0:4])          # renormalize q

    # wheel speed saturation
    Omega = x[7:7+Nw]
    if np.any(np.abs(Omega) > Omega_max):
        if not sat_warned:
            print("[warning] wheel speed saturated -> momentum dumping (desaturation) needed")
            sat_warned = True
        x[7:7+Nw] = np.clip(Omega, -Omega_max, Omega_max)

    t += dt_state

# =========================================
# Plots
# =========================================
plt.figure(figsize=(7, 3))
plt.plot(t_history, q_history)
plt.plot(t_history, np.linalg.norm(q_history, axis=1))
plt.axhline(q_set[0], color='gray', ls=':', lw=1, alpha=0.6)
plt.legend(["q0","q1","q2","q3","|q|"])
plt.title("Quaternion  (- wheels-only, controller @ 1 Hz)")
plt.xlabel("Time [s]"); plt.grid(); plt.show()

plt.figure(figsize=(7, 3))
plt.plot(t_history, qerr_history)
plt.legend(["e1","e2","e3"])
plt.title("Attitude error  (vector part of delta-q)")
plt.xlabel("Time [s]"); plt.grid(); plt.show()

plt.figure(figsize=(7, 3))
plt.plot(t_history, w_history * 180/np.pi)
plt.legend(["wx","wy","wz"])
plt.title("Angular Velocity [deg/s]  (wheels-only)")
plt.xlabel("Time [s]"); plt.grid(); plt.show()

plt.figure(figsize=(7, 3))
plt.plot(t_history, Omega_history * 60/(2*np.pi))
plt.axhline( Omega_max * 60/(2*np.pi), color='r', ls=':', lw=2, alpha=0.5)
plt.axhline(-Omega_max * 60/(2*np.pi), color='r', ls=':', lw=2, alpha=0.5)
plt.axvline(tf, color='k', ls=':', lw=1, alpha=0.6)
plt.legend([f"RW{i+1}" for i in range(Nw)])
plt.title("Wheel speed [rpm]  (red = saturation)")
plt.xlabel("Time [s]"); plt.grid(); plt.show()

plt.figure(figsize=(7, 3))
for i in range(Nw):
    plt.step(t_history, tauwcmd_history[:, i] * 1e3, where='post', label=f"cmd RW{i+1}")
plt.plot(t_history, tauw_history * 1e3, '--', alpha=0.7)
plt.legend()
plt.title("Wheel torque [mN.m]  (-- actual servo,  step = 1 Hz command)")
plt.xlabel("Time [s]"); plt.grid(); plt.show()

plt.figure(figsize=(7, 3))
plt.plot(t_history, taurw_history * 1e6)
plt.legend(["tau_x","tau_y","tau_z"])
plt.title("Control torque on body from wheels [uN.m]")
plt.xlabel("Time [s]"); plt.grid(); plt.show()

plt.figure(figsize=(7, 3))
plt.plot(t_history, hw_history * 1e3)
plt.plot(t_history, np.linalg.norm(hw_history, axis=1) * 1e3, 'k--')
plt.legend(["hx","hy","hz","|h|"])
plt.title("Wheel angular momentum [mN.m.s]")
plt.xlabel("Time [s]"); plt.grid(); plt.show()
```

# Closed loop (VAT only, no COGU)

```python
# ============================================================================
#  CLOSED-LOOP VAT ATTITUDE CONTROL  —  classic baseline (VAT ONLY)
# ----------------------------------------------------------------------------
#  Head-to-head against your COGU open-loop optimal pulse train.
#
#  WHY THE STIFF PD DID NOT CONVERGE
#  ---------------------------------
#  With your hardware each VAT delivers only ~3.7 uN.m (|r x f| ~ 0.079,
#  F_max = 50 uN), and the pairing {1,2},{3,4} forbids adding two thrusters
#  on the same axis. So the x-axis angular acceleration tops out at
#       alpha_x ~ 3.7e-6 / J1 ~ 1.1e-4 rad/s^2,
#  while a 180 deg rest-to-rest slew in tf=300 s needs ~1.4e-4 rad/s^2.
#  A high-gain PD therefore demands ~1000x the available torque, saturates
#  permanently and limit-cycles. That is exactly the gap COGU exploits.
#
#  CLASSIC FIX (still a textbook controller)
#  -----------------------------------------
#  Rate-limited eigenaxis control (sliding surface):
#     w_cmd   = sat( -K_att * att_err , w_lim )     # max slew rate it can brake
#     tau_des = -Kd * (w - w_cmd)                    # saturating accel/brake
#  w_lim and K_att are sized FROM the measured actuator authority so the
#  controller never commands a rate it cannot stop in the remaining angle.
#  VAT only: reaction wheels stay inert (tau_w_cmd = 0).
#
#  Assumes from your guidance solve:
#     Jsat_double, x_value, u_value, T, tau_double, tf, Fth_max_double,
#     rth1_double..rth4_double, fth1_double..fth4_double
#  and helpers: quat_mult, quat_conj, quat_rotate, dynamics, rk4_step
# ============================================================================

import numpy as np
import matplotlib.pyplot as plt

# =========================================
# Thruster control matrix B   (M = sum_i u_i (rth_i x fth_i))
# =========================================
B = np.column_stack([
    np.cross(rth1_double[:, 0], fth1_double[:, 0]),
    np.cross(rth2_double[:, 0], fth2_double[:, 0]),
    np.cross(rth3_double[:, 0], fth3_double[:, 0]),
    np.cross(rth4_double[:, 0], fth4_double[:, 0]),
])

# =========================================
# Perturbations / uncertainty  (SAME seed as the COGU run -> fair compare)
# =========================================
pert_seed = 0
rng = np.random.default_rng(pert_seed)

thrust_nominal    = Fth_max_double
thrust_offset     = 0.0
thrust_sigma      = Fth_max_double / 10
thrust_bias_sigma = Fth_max_double / 20
thrust_bias       = thrust_bias_sigma * rng.standard_normal(4)

inertia_var = 0.50    # +/- 50% multiplicative inertia error (true plant vs model)
tau_dist = np.array([1.0e-7, -0.5e-7, 0.8e-7])   # [N.m] residual torque (0 to disable)  (TUNE)

F_max = thrust_nominal


def apply_thruster_noise(u_cmd):
    fired = u_cmd > 0.0
    noisy = u_cmd + thrust_offset + thrust_bias + thrust_sigma * rng.standard_normal(4)
    return np.where(fired, np.clip(noisy, 0.0, None), 0.0)


# =========================================
# Reaction wheels: present in the state but INERT (no command)
# =========================================
Aw        = np.eye(3)
Nw        = Aw.shape[1]
Aw_pinv   = np.eye(3)
Jw        = 4.774648292756861e-06
tau_w_max = 1.0e-3
Omega_max = 8000.0 * (2 * np.pi / 60)
tau_motor = 0.10

# =========================================
# PWPF modulator  (classic on/off thruster modulation)
# =========================================
MODULATOR = "pwpf"     # "pwpf" (classic) | "schmitt" (plain relay)
Km   = 5.0
Tm   = 0.50            # a bit slower -> less chatter   (TUNE)
Uon  = 0.45
Uoff = 0.15


def pwpf_update(f, fire, c, dt):
    e = c - fire
    f = f + dt * (Km * e - f) / Tm
    new = fire.copy()
    new[f >= Uon] = 1.0
    new[f <= Uoff] = 0.0
    return f, new


def schmitt_update(fire, c):
    new = fire.copy()
    new[c >= Uon] = 1.0
    new[c <= Uoff] = 0.0
    return new


# =========================================
# VAT allocation: desired body torque -> per-thruster duty in [0,1]
# =========================================
def allocate_vat(tau_des):
    duty = np.zeros(4)
    for i in range(4):
        di = B[:, i]
        proj = np.dot(tau_des, di) / (np.dot(di, di) + 1e-30)
        duty[i] = proj / F_max
    duty = np.clip(duty, 0.0, 1.0)
    if duty[0] >= duty[1]: duty[1] = 0.0
    else:                  duty[0] = 0.0
    if duty[2] >= duty[3]: duty[3] = 0.0
    else:                  duty[2] = 0.0
    return duty


# =========================================
# Actuator-authority diagnostics  ->  used to size the controller
# =========================================
I_nominal = Jsat_double.copy()
d_norm     = np.linalg.norm(B, axis=0)                 # |r x f| per thruster
tau_1th    = F_max * d_norm.max()                      # best single-thruster torque
# per-axis best single-thruster torque / inertia  -> available angular accel
tau_axis   = F_max * np.max(np.abs(B), axis=1)         # best |torque| per body axis
alpha_axis = tau_axis / np.diag(I_nominal)
alpha_x    = alpha_axis[0]                             # maneuver is ~about x

print("===============  ACTUATOR AUTHORITY  ===============")
print(f"  max torque / thruster : {tau_1th*1e6:.3f} uN.m")
print(f"  per-axis torque [uN.m]: {np.round(tau_axis*1e6,3)}")
print(f"  per-axis accel [rad/s2]: {np.round(alpha_axis,6)}")
req_alpha = 4*np.pi/tf**2
print(f"  alpha needed for 180deg/{tf:.0f}s : {req_alpha:.2e}  (vs avail x {alpha_x:.2e})")
print("====================================================")

# ----- size the rate-limited (sliding) controller from alpha_x -----
brake_angle = np.deg2rad(25.0)                          # angle reserved to brake (TUNE)
w_lim   = np.sqrt(2.0 * alpha_x * brake_angle)          # cruise rate it can stop in time
w_lim   = min(w_lim, omegab_max_double * 0.9)           # respect the rate bound
K_att   = w_lim / np.sin(brake_angle / 2.0)             # start braking at brake_angle
Kd      = 2.0e-3                                        # high enough to saturate accel/brake (TUNE)
print(f"  controller: w_lim={np.rad2deg(w_lim):.3f} deg/s, K_att={K_att:.3f}, Kd={Kd:.1e}")

# =========================================
# Setup
# =========================================
while True:                                             # TRUE perturbed inertia (PD blind to it)
    P = 1.0 + inertia_var * (2.0 * rng.random((3, 3)) - 1.0)
    P = 0.5 * (P + P.T)
    I_true = I_nominal * P
    if np.all(np.linalg.eigvalsh(I_true) > 0.0):
        break

params = {
    "I": I_true, "I_inv": np.linalg.inv(I_true),
    "B": B, "Kp": np.zeros((3, 3)), "Kd": np.zeros((3, 3)),
    "Aw": Aw, "Aw_pinv": Aw_pinv, "Jw": Jw,
    "tau_w_max": tau_w_max, "tau_motor": tau_motor,
    "use_thrusters": True,
}

q_set = end_quat[0:4, 0].copy(); q_set /= np.linalg.norm(q_set)
w_set = np.zeros(3)

x = np.concatenate((start_quat[0:4, 0].copy(),x_value[4:7, 0].copy(), np.zeros(Nw), np.zeros(Nw)))

dt_state  = 0.1
hold_time = 1500.0          # generous: the weak-actuator slew is SLOW (TUNE)
sim_time  = tf + hold_time
N = int(round(sim_time / dt_state)) + 1

tau_w_cmd = np.zeros(Nw)   # wheels inert

q_history    = np.zeros((N, 4))
w_history    = np.zeros((N, 3))
u_history    = np.zeros((N, 4))
ucmd_history = np.zeros((N, 4))
duty_history = np.zeros((N, 4))
errang_hist  = np.zeros(N)
t_history    = np.zeros(N)

f_pwpf    = np.zeros(4)
fire      = np.zeros(4)
fire_prev = np.zeros(4)
rising_edges = np.zeros(4)

t = 0.0

# =========================================
# Simulation loop  (rate-limited VAT-only closed loop)
# =========================================
for i in range(N):
    q = x[0:4]; w = x[4:7]
    q_history[i] = q; w_history[i] = w
    t_history[i] = t

    q_err = quat_mult(quat_conj(q_set), q)
    if q_err[0] < 0.0:
        q_err = -q_err
    att_err = q_err[1:4]
    errang_hist[i] = 2.0 * np.degrees(np.arccos(np.clip(abs(q_err[0]), 0.0, 1.0)))

    # ---- rate-limited eigenaxis (sliding) law ----
    w_cmd = -K_att * att_err
    nrm = np.linalg.norm(w_cmd)
    if nrm > w_lim:
        w_cmd *= w_lim / nrm
    tau_des = -Kd * (w - w_cmd)

    # ---- VAT: allocate -> modulate -> on/off ----
    duty = allocate_vat(tau_des)
    duty_history[i] = duty
    if MODULATOR == "pwpf":
        f_pwpf, fire = pwpf_update(f_pwpf, fire, duty, dt_state)
    else:
        fire = schmitt_update(fire, duty)

    rising_edges += np.logical_and(fire == 1.0, fire_prev == 0.0)
    fire_prev = fire.copy()

    u_cmd          = fire * F_max
    u_current      = apply_thruster_noise(u_cmd)
    ucmd_history[i] = u_cmd
    u_history[i]    = u_current

    x = rk4_step(dynamics, t, x, u_current, tau_w_cmd, dt_state, params)
    x[0:4] /= np.linalg.norm(x[0:4])
    t += dt_state


# =========================================
# Metrics  +  comparison vs COGU
# =========================================
def settling_time(err_deg, t_hist, tol_deg=1.0):
    below = err_deg <= tol_deg
    for k in range(len(below)):
        if below[k] and np.all(below[k:]):
            return t_hist[k]
    return np.nan

ts          = settling_time(errang_hist, t_history, tol_deg=1.0)
ss_err      = np.mean(errang_hist[t_history >= t_history[-1] - 60.0])
on_fraction = (u_history > 0.0).mean(axis=0)
total_imp   = np.sum(u_history) * dt_state

cogu_fire  = (u_value > thrust_nominal / 2).astype(float)
cogu_edges = np.sum((cogu_fire[:, 1:] == 1) & (cogu_fire[:, :-1] == 0), axis=1) \
             + (cogu_fire[:, 0] == 1)

print("\n=========  CLOSED-LOOP VAT ONLY (rate-limited classic)  =========")
print(f"  modulator            : {MODULATOR}")
print(f"  settling (<1 deg)    : {ts:.1f} s" if not np.isnan(ts) else "  settling (<1 deg)    : not reached")
print(f"  (COGU finishes at tf = {tf:.0f} s)")
print(f"  steady-state error   : {ss_err:.3f} deg")
print(f"  VAT firings (edges)  : {rising_edges.astype(int)}  total={int(rising_edges.sum())}")
print(f"  VAT on-time fraction : {np.round(on_fraction,3)}")
print(f"  total impulse [N.s]  : {total_imp:.3e}")
print("-----------------------  COGU (open-loop)  ----------------------")
print(f"  VAT firings (edges)  : {cogu_edges.astype(int)}  total={int(cogu_edges.sum())}")
print("=================================================================\n")


# =========================================
# Plots
# =========================================
plt.figure(figsize=(7, 3))
plt.plot(t_history, errang_hist)
plt.axhline(1.0, color='g', ls=':', lw=1, label='1 deg')
plt.axvline(tf, color='k', ls=':', lw=1, alpha=0.6, label='tf (COGU done)')
plt.title("Pointing error vs target  (rate-limited VAT only)")
plt.xlabel("Time [s]"); plt.ylabel("error [deg]")
plt.legend(); plt.grid(); plt.show()

plt.figure(figsize=(7, 3))
plt.plot(t_history, q_history)
plt.plot(t_history, np.linalg.norm(q_history, axis=1), 'k--', lw=0.8)
for j in range(4):
    plt.axhline(q_set[j], color='gray', ls=':', lw=0.9)
plt.axvline(tf, color='k', ls=':', lw=1, alpha=0.6)
plt.legend(["q0", "q1", "q2", "q3", "|q|"])
plt.title("Quaternion  (-- = target qf)")
plt.xlabel("Time [s]"); plt.grid(); plt.show()

plt.figure(figsize=(7, 3))
plt.plot(t_history, w_history * 180 / np.pi)
plt.axhline( np.rad2deg(w_lim), color='m', ls=':', lw=1, alpha=0.6, label='w_lim')
plt.axhline(-np.rad2deg(w_lim), color='m', ls=':', lw=1, alpha=0.6)
plt.axvline(tf, color='k', ls=':', lw=1, alpha=0.6)
plt.legend(["wx", "wy", "wz", "w_lim"])
plt.title("Angular velocity [deg/s]")
plt.xlabel("Time [s]"); plt.grid(); plt.show()

fig, axs = plt.subplots(4, 1, sharex=True, figsize=(6, 6))
colors = ['tab:blue', 'tab:orange', 'tab:green', 'tab:red']
for k in range(4):
    axs[k].step(t_history, u_history[:, k] * 1e6, where='post',
                color=colors[k], label=f'VAT{k+1} (noisy)')
    axs[k].step(t_history, ucmd_history[:, k] * 1e6, where='post',
                color='k', lw=0.8, alpha=0.4, label='ideal')
    axs[k].set_ylabel(f'F{k+1} [uN]'); axs[k].grid(True)
    axs[k].legend(loc='upper right', fontsize=7)
axs[-1].set_xlabel('Time [s]')
fig.suptitle('VAT firing sequence  (closed-loop PWPF)')
fig.tight_layout(); plt.show()
```
